# Proyecto: Modelo minimal de potenciales de acción

### 1. Escriba ecuaciones para $w$ y $v$ con el número menor de términos posible, y tal que pueda producir potenciales de acción. 

**Pista:** intente con sólo dos corrientes. ¿Por qué? Ambas deben ser no inactivantes.

**Objetivo**

Queremos modelar una membrana excitable usando dos variables:

$$
v(t) = \text{potencial transmembranal},
$$

$$
w(t) = \text{variable de recuperación}.
$$

La variable $v$ describe el voltaje de la membrana y la variable $w$ describe, de manera simplificada, la proporción de canales de potasio abiertos o, más generalmente, la activación de una corriente que baja el voltaje.

La idea fundamental es que un potencial de acción necesita dos procesos:

$$
\text{subida rápida del voltaje}
+
\text{bajada retardada del voltaje}.
$$

Por eso buscamos un modelo con dos corrientes:

1. Una corriente ascendente, que empuje el voltaje hacia valores altos.
2. Una corriente descendente, que empuje el voltaje hacia valores bajos.

Llamaremos a estas corrientes:

$$
J_U = \text{corriente Up},
$$

$$
J_D = \text{corriente Down}.
$$

La corriente $J_U$ representa una corriente despolarizante, por ejemplo de sodio o calcio. La corriente $J_D$ representa una corriente repolarizante, por ejemplo de potasio.

**Intuición física y biofísica**

La membrana celular puede pensarse como un capacitor. En electrónica, un capacitor satisface

$$
C\frac{dV}{dt}=I.
$$

Esto significa que el voltaje cambia cuando hay una corriente neta cargando o descargando el capacitor.

En una membrana neuronal ocurre algo análogo. La bicapa lipídica funciona como un capacitor, mientras que los canales iónicos funcionan como ramas conductivas en paralelo. Cada rama iónica puede pensarse como una conductancia conectada a una batería, cuyo voltaje corresponde al potencial de reversa del ion.

Así, una corriente iónica suele escribirse como

$$
I = g(V-E),
$$

donde:

- $g$ es una conductancia,
- $V$ es el potencial de membrana,
- $E$ es el potencial de reversa.

En nuestro caso, después de dividir por la capacitancia de membrana, trabajamos con corrientes normalizadas. Por eso escribimos ecuaciones directamente para $\dot v$.

La forma conceptual es:

$$
\dot v = \text{corriente externa} - \text{corrientes iónicas}.
$$

El modelo minimal que queremos construir debe contener lo necesario para producir un pulso:

- una corriente que suba rápidamente el voltaje,
- una corriente que lo baje después de cierto retraso.

**¿Por qué dos corrientes?**

Un potencial de acción tiene una forma característica:

1. El voltaje está cerca del reposo.
2. Un estímulo lo eleva.
3. Si cruza cierto umbral, sube rápidamente.
4. Después alcanza un máximo.
5. Luego baja rápidamente.
6. Finalmente regresa al reposo.

Para obtener esta forma necesitamos dos fuerzas opuestas.

**Corriente ascendente**

La corriente ascendente debe empujar el voltaje hacia un valor alto $v_U$. Este valor puede interpretarse como un potencial de reversa tipo sodio o calcio.

Si $v_U$ es alto y $v<v_U$, entonces la corriente asociada a $v_U$ tiende a subir el voltaje.

**Corriente descendente**

La corriente descendente debe empujar el voltaje hacia un valor bajo $v_D$. Este valor puede interpretarse como un potencial de reversa tipo potasio.

Si $v_D$ es bajo y $v>v_D$, entonces la corriente asociada a $v_D$ tiende a bajar el voltaje.

Por eso asumimos

$$
v_D < v < v_U.
$$

**¿Por qué las corrientes deben ser no inactivantes?**

En modelos más detallados, como Hodgkin-Huxley, la corriente de sodio tiene una variable de activación y una variable de inactivación. Eso permite describir con más precisión cómo se encienden y se apagan los canales de sodio.

Sin embargo, agregar una variable de inactivación haría que el modelo dejara de ser minimal. En vez de tener sólo dos variables, tendríamos al menos tres:

$$
(v,w,h),
$$

donde $h$ sería una variable de inactivación.

Como el proyecto pide usar el menor número de términos posible, evitamos introducir inactivación. En su lugar, hacemos que la corriente ascendente se apague de manera indirecta:

1. cuando $v$ se acerca a $v_U$, su fuerza impulsora disminuye;
2. cuando $w$ aumenta, la corriente descendente se vuelve más fuerte;
3. la corriente descendente termina dominando y baja el voltaje.

Así, no necesitamos una variable de inactivación. La recuperación del potencial de acción se produce por la activación retardada de $w$.


**Construcción del modelo**

Proponemos el siguiente modelo minimal:

$$
\dot v = J_F - J_U(v) - J_D(v,w),
$$

$$
\dot w = \alpha(v)(1-w)-\beta(v)w.
$$

Aquí:

- $v$ es el potencial transmembranal,
- $w$ es la variable de recuperación,
- $J_F$ es una corriente externa o forzamiento,
- $J_U$ es una corriente ascendente,
- $J_D$ es una corriente descendente.

Definimos la corriente ascendente como

$$
J_U(v)=g_U S_m(v)(v-v_U),
$$

y la corriente descendente como

$$
J_D(v,w)=g_D w(v-v_D).
$$

Por tanto, el sistema queda:

$$
\boxed{
\dot v
=
J_F
-
g_U S_m(v)(v-v_U)
-
g_D w(v-v_D)
}
$$

$$
\boxed{
\dot w
=
\alpha(v)(1-w)-\beta(v)w
}
$$


La función $S_m(v)$ describe qué tan activa está la corriente ascendente. Usamos una función sigmoidal:

$$
\boxed{
S_m(v)
=
\frac{1}{1+\exp[s_m(v_m-v)]}
}
$$

Esta función tiene las siguientes propiedades:

- Si $v$ es muy bajo, entonces $S_m(v)\approx 0$.
- Si $v$ es muy alto, entonces $S_m(v)\approx 1$.
- Si $v=v_m$, entonces

$$
S_m(v_m)=\frac{1}{2}.
$$

Por eso $v_m$ se interpreta como el potencial de media activación.

El parámetro $s_m$ controla qué tan abrupta es la activación. Si $s_m$ es grande, la activación cambia bruscamente cerca de $v_m$. Si $s_m$ es pequeño, la transición es más gradual.

**Dinámica de la variable de recuperación**

La variable $w$ satisface

$$
\dot w = \alpha(v)(1-w)-\beta(v)w.
$$

Esta ecuación tiene una interpretación sencilla:

- $\alpha(v)(1-w)$ abre canales o activa la variable $w$,
- $\beta(v)w$ cierra canales o desactiva la variable $w$.

Expandiendo:

$$
\dot w
=
\alpha(v)-\alpha(v)w-\beta(v)w.
$$

Agrupando términos:

$$
\dot w
=
\alpha(v)-[\alpha(v)+\beta(v)]w.
$$

Definimos

$$
\rho(v)=\alpha(v)+\beta(v).
$$

Entonces

$$
\dot w
=
\alpha(v)-\rho(v)w.
$$

Factorizando:

$$
\dot w
=
\rho(v)
\left(
\frac{\alpha(v)}{\rho(v)}-w
\right).
$$

Definimos el estado estacionario de $w$ como

$$
S_w(v)=\frac{\alpha(v)}{\alpha(v)+\beta(v)}.
$$

Así,

$$
\boxed{
\dot w
=
\rho(v)(S_w(v)-w)
}
$$

Esto significa que $w$ intenta seguir a $S_w(v)$, pero no lo hace instantáneamente. La rapidez con la que se acerca está determinada por $\rho(v)$.

Por lo tanto, $w$ es una variable lenta o de recuperación.

**Elección de las tasas $\alpha(v)$ y $\beta(v)$**

Tomamos

$$
\boxed{
\alpha(v)=r\exp[s_wb_w(v-v_w)]
}
$$

y

$$
\boxed{
\beta(v)=r\exp[s_w(b_w-1)(v-v_w)].
}
$$

Aquí:

- $r$ controla la escala temporal de la variable $w$,
- $v_w$ es el potencial de media activación de $w$,
- $s_w$ controla la pendiente de activación,
- $b_w$ controla el sesgo entre activación y desactivación.

Con estas definiciones,

$$
S_w(v)
=
\frac{\alpha(v)}{\alpha(v)+\beta(v)}.
$$

Calculemos explícitamente su forma.

Primero:

$$
S_w(v)
=
\frac{1}{1+\frac{\beta(v)}{\alpha(v)}}.
$$

Ahora,

$$
\frac{\beta(v)}{\alpha(v)}
=
\frac{
r\exp[s_w(b_w-1)(v-v_w)]
}{
r\exp[s_wb_w(v-v_w)]
}.
$$

Cancelando $r$:

$$
\frac{\beta(v)}{\alpha(v)}
=
\exp\left(
s_w(b_w-1)(v-v_w)-s_wb_w(v-v_w)
\right).
$$

Simplificando el exponente:

$$
s_w(b_w-1)(v-v_w)-s_wb_w(v-v_w)
=
-s_w(v-v_w).
$$

Por tanto,

$$
\frac{\beta(v)}{\alpha(v)}
=
\exp[-s_w(v-v_w)].
$$

Equivalente a

$$
\frac{\beta(v)}{\alpha(v)}
=
\exp[s_w(v_w-v)].
$$

Así,

$$
\boxed{
S_w(v)
=
\frac{1}{1+\exp[s_w(v_w-v)]}
}
$$

Esta función tiene la misma forma funcional que $S_m(v)$.

**Revisión de signos**

El sistema es

$$
\dot v
=
J_F
-
g_U S_m(v)(v-v_U)
-
g_D w(v-v_D).
$$

Asumimos

$$
v_D<v<v_U.
$$

Analicemos primero la corriente ascendente:

$$
J_U(v)=g_U S_m(v)(v-v_U).
$$

Como $g_U>0$, $S_m(v)>0$ y $v-v_U<0$, se tiene

$$
J_U(v)<0.
$$

Pero en la ecuación aparece como $-J_U(v)$. Entonces

$$
-J_U(v)>0.
$$

Por tanto, $J_U$ contribuye a subir el voltaje:

$$
\boxed{
J_U<0
\quad\Rightarrow\quad
-J_U>0
\quad\Rightarrow\quad
\dot v \text{ aumenta}.
}
$$

Ahora analicemos la corriente descendente:

$$
J_D(v,w)=g_Dw(v-v_D).
$$

Como $g_D>0$, $w\geq 0$ y $v-v_D>0$, se tiene

$$
J_D(v,w)>0.
$$

Pero aparece como $-J_D(v,w)$. Entonces

$$
-J_D(v,w)<0.
$$

Por tanto, $J_D$ contribuye a bajar el voltaje:

$$
\boxed{
J_D>0
\quad\Rightarrow\quad
-J_D<0
\quad\Rightarrow\quad
\dot v \text{ disminuye}.
}
$$

Esta es la razón por la que llamamos a $J_U$ corriente `Up` y a $J_D$ corriente `Down`.

**Unidades**

Tomamos:

$$
[v]=\mathrm{mV},
$$

$$
[t]=\mathrm{ms},
$$

$$
[w]=1.
$$

Entonces

$$
[\dot v]=\mathrm{mV/ms}.
$$

Las corrientes normalizadas $J_F$, $J_U$ y $J_D$ tienen unidades de $\mathrm{mV/ms}$.

Como

$$
J_U(v)=g_U S_m(v)(v-v_U),
$$

y $S_m(v)$ es adimensional, se necesita que

$$
[g_U]=\mathrm{ms}^{-1}.
$$

De manera similar,

$$
J_D(v,w)=g_Dw(v-v_D),
$$

y como $w$ es adimensional,

$$
[g_D]=\mathrm{ms}^{-1}.
$$

En la ecuación de $w$,

$$
\dot w=\alpha(v)(1-w)-\beta(v)w,
$$

como $w$ es adimensional,

$$
[\dot w]=\mathrm{ms}^{-1}.
$$

Por lo tanto,

$$
[\alpha(v)]=[\beta(v)]=\mathrm{ms}^{-1}.
$$

Como $\alpha$ y $\beta$ contienen el parámetro $r$, se tiene

$$
[r]=\mathrm{ms}^{-1}.
$$

Los exponentes deben ser adimensionales. Por eso,

$$
[s_m]=[s_w]=\mathrm{mV}^{-1}.
$$

**¿Por qué este modelo puede producir potenciales de acción?**

El modelo puede producir potenciales de acción porque combina dos mecanismos esenciales:

$$
\text{retroalimentación positiva rápida}
+
\text{retroalimentación negativa retardada}.
$$

#### Fase 1: reposo

El sistema comienza cerca de un estado de reposo. El voltaje $v$ es bajo y la variable $w$ también es relativamente baja.

#### Fase 2: estímulo

Una corriente externa $J_F$ o una perturbación inicial aumenta $v$.

#### Fase 3: activación rápida de la corriente ascendente

Al subir $v$, aumenta $S_m(v)$.

Entonces aumenta la magnitud de la corriente ascendente:

$$
J_U(v)=g_US_m(v)(v-v_U).
$$

Como $-J_U>0$, esto empuja $v$ hacia arriba.

Tenemos entonces una retroalimentación positiva:

$$
v\uparrow
\quad\Rightarrow\quad
S_m(v)\uparrow
\quad\Rightarrow\quad
-J_U\uparrow
\quad\Rightarrow\quad
v\uparrow.
$$

Esto produce la fase rápida de subida del potencial de acción.

#### Fase 4: activación retardada de la corriente descendente

Mientras $v$ sube, la variable $w$ también comienza a aumentar, pero no instantáneamente. Esto ocurre porque $w$ satisface una ecuación diferencial de primer orden.

Cuando $w$ crece, aumenta la corriente descendente:

$$
J_D(v,w)=g_Dw(v-v_D).
$$

Como $-J_D<0$, esta corriente empieza a frenar la subida de $v$.

#### Fase 5: repolarización

Cuando la corriente descendente domina sobre la corriente ascendente, se tiene

$$
\dot v<0.
$$

Entonces el voltaje comienza a bajar.

#### Fase 6: recuperación

Al bajar $v$, la activación de $w$ disminuye y el sistema regresa lentamente hacia el reposo.

La trayectoria completa es:

$$
\text{reposo}
\to
\text{subida rápida}
\to
\text{pico}
\to
\text{repolarización}
\to
\text{recuperación}.
$$

Esto corresponde a un potencial de acción.

**Modelo propuesto**

El modelo minimal propuesto es:

$$
\boxed{
\begin{aligned}
\dot v
&=
J_F
-
g_U S_m(v)(v-v_U)
-
g_D w(v-v_D),
\\[4pt]
\dot w
&=
\alpha(v)(1-w)-\beta(v)w.
\end{aligned}
}
$$

con

$$
\boxed{
S_m(v)
=
\frac{1}{1+\exp[s_m(v_m-v)]}
}
$$

$$
\boxed{
\alpha(v)
=
r\exp[s_wb_w(v-v_w)]
}
$$

$$
\boxed{
\beta(v)
=
r\exp[s_w(b_w-1)(v-v_w)]
}
$$

y, equivalentemente,

$$
\boxed{
S_w(v)
=
\frac{\alpha(v)}{\alpha(v)+\beta(v)}
=
\frac{1}{1+\exp[s_w(v_w-v)]}.
}
$$

### 2. Calcule soluciones que validen el modelo con los requisitos de arriba.

Para validar el modelo minimal, se resolvió numéricamente el sistema

$$
\dot v
=
J_F
-
g_U S_m(v)(v-v_U)
-
g_D w(v-v_D),
$$

$$
\dot w
=
\alpha(v)(1-w)-\beta(v)w.
$$

Se usaron los parámetros

$$
g_U=2,\quad g_D=10,\quad r=0.1,
$$

$$
v_U=60,\quad v_D=-80,\quad v_m=-20,\quad v_w=0,
$$

$$
s_m=0.22,\quad s_w=0.11,\quad b_w=0.35.
$$

Se aplicó un pulso breve de corriente externa

$$
J_{\text{stim}}(t)
=
\begin{cases}
100, & 1\leq t\leq 1.5,\\
0, & \text{en otro caso}.
\end{cases}
$$

La solución obtenida muestra una excursión rápida del potencial transmembranal seguida de repolarización. Para evitar contar el estímulo como parte de la velocidad intrínseca del potencial de acción, las mediciones se realizaron después de apagar el pulso.

Se obtuvo aproximadamente

$$
\max \dot v = 113.62 \ \mathrm{mV/ms},
$$

y, tomando como umbral operacional

$$
v_\theta=-20 \ \mathrm{mV},
$$

la duración del potencial de acción fue

$$
T_{PA}=2.49 \ \mathrm{ms}.
$$

Por lo tanto, el modelo satisface los requisitos del proyecto:

$$
\max \dot v>100 \ \mathrm{mV/ms},
$$

y

$$
2 \ \mathrm{ms}<T_{PA}<3 \ \mathrm{ms}.
$$

La solución valida que el modelo minimal de dos corrientes no inactivantes puede producir un potencial de acción. La corriente ascendente genera la despolarización rápida, mientras que la variable lenta $w$ activa la corriente descendente y produce la repolarización.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp
from scipy.optimize import brentq
```

```python
# Parámetros del modelo minimal Up-Down

parametros = {
    "gU": 2.0,
    "gD": 10.0,
    "r": 0.1,
    "vU": 60.0,
    "vD": -80.0,
    "vm": -20.0,
    "vw": 0.0,
    "sm": 0.22,
    "sw": 0.11,
    "bw": 0.35,
    "JF": 0.0
}
```

```python
def construir_funciones(parametros):
    """
    Construye las funciones del modelo minimal Up-Down.
    """

    gU = parametros["gU"]
    gD = parametros["gD"]
    r = parametros["r"]
    vU = parametros["vU"]
    vD = parametros["vD"]
    vm = parametros["vm"]
    vw = parametros["vw"]
    sm = parametros["sm"]
    sw = parametros["sw"]
    bw = parametros["bw"]
    JF = parametros["JF"]

    def Sm(v):
        return 1.0 / (1.0 + np.exp(sm * (vm - v)))

    def alpha(v):
        return r * np.exp(sw * bw * (v - vw))

    def beta(v):
        return r * np.exp(sw * (bw - 1.0) * (v - vw))

    def rho(v):
        return alpha(v) + beta(v)

    def Sw(v):
        return alpha(v) / rho(v)

    def JU(v):
        return gU * Sm(v) * (v - vU)

    def JD(v, w):
        return gD * w * (v - vD)

    def campo(t, y, corriente_externa):
        v, w = y

        dv = JF + corriente_externa(t) - JU(v) - JD(v, w)
        dw = alpha(v) * (1.0 - w) - beta(v) * w

        return np.array([dv, dw])

    return Sm, alpha, beta, rho, Sw, JU, JD, campo
```

```python
def encontrar_puntos_fijos(parametros, numero_puntos=2000):
    """
    Encuentra puntos fijos resolviendo la ecuación de equilibrio en v.
    """

    Sm, alpha, beta, rho, Sw, JU, JD, campo = construir_funciones(parametros)

    vD = parametros["vD"]
    vU = parametros["vU"]
    JF = parametros["JF"]

    def ecuacion_equilibrio(v):
        return JF - JU(v) - JD(v, Sw(v))

    malla_v = np.linspace(vD, vU, numero_puntos)
    valores = ecuacion_equilibrio(malla_v)

    raices = []

    for i in range(len(malla_v) - 1):
        if valores[i] * valores[i + 1] < 0:
            raiz = brentq(ecuacion_equilibrio, malla_v[i], malla_v[i + 1])
            raices.append(raiz)

    puntos_fijos = [(v, Sw(v)) for v in raices]

    return puntos_fijos
```

```python
puntos_fijos = encontrar_puntos_fijos(parametros)

for i, (v, w) in enumerate(puntos_fijos):
    print(f"Punto fijo {i+1}: v* = {v:.4f} mV, w* = {w:.6f}")
```

```python
def simular_con_pulso(
    parametros,
    amplitud_pulso=100.0,
    inicio_pulso=1.0,
    duracion_pulso=0.5,
    tmax=30.0,
    paso_maximo=0.001
):
    """
    Simula el modelo minimal Up-Down con un pulso breve de corriente externa.
    """

    Sm, alpha, beta, rho, Sw, JU, JD, campo = construir_funciones(parametros)

    puntos_fijos = encontrar_puntos_fijos(parametros)
    v_reposo, w_reposo = puntos_fijos[0]

    def corriente_externa(t):
        if inicio_pulso <= t <= inicio_pulso + duracion_pulso:
            return amplitud_pulso
        else:
            return 0.0

    def sistema(t, y):
        return campo(t, y, corriente_externa)

    solucion = solve_ivp(
        sistema,
        [0.0, tmax],
        [v_reposo, w_reposo],
        max_step=paso_maximo,
        rtol=1e-8,
        atol=1e-10
    )

    t = solucion.t
    v = solucion.y[0]
    w = solucion.y[1]

    dv = np.array([sistema(ti, yi)[0] for ti, yi in zip(t, solucion.y.T)])
    I = np.array([corriente_externa(ti) for ti in t])

    return t, v, w, dv, I
```

```python
t, v, w, dv, I = simular_con_pulso(parametros)
```

```python
def duracion_por_cruces(t, v, umbral):
    """
    Calcula la duración entre el cruce ascendente y descendente
    de un umbral dado.
    """

    arriba = v >= umbral

    if not np.any(arriba):
        return np.nan, np.nan, np.nan

    indices = np.where(arriba)[0]

    i_inicio = indices[0]
    i_final = indices[-1]

    # Cruce ascendente interpolado
    t_subida = t[i_inicio]

    if i_inicio > 0:
        t0, t1 = t[i_inicio - 1], t[i_inicio]
        v0, v1 = v[i_inicio - 1], v[i_inicio]

        if v1 != v0:
            t_subida = t0 + (umbral - v0) * (t1 - t0) / (v1 - v0)

    # Cruce descendente interpolado
    t_bajada = t[i_final]

    if i_final < len(t) - 1:
        t0, t1 = t[i_final], t[i_final + 1]
        v0, v1 = v[i_final], v[i_final + 1]

        if v1 != v0:
            t_bajada = t0 + (umbral - v0) * (t1 - t0) / (v1 - v0)

    duracion = t_bajada - t_subida

    return duracion, t_subida, t_bajada
```

```python
def medir_potencial_de_accion(
    t,
    v,
    w,
    dv,
    inicio_pulso=1.0,
    duracion_pulso=0.5,
    umbral=-20.0
):
    """
    Mide las características del potencial de acción después del estímulo.
    """

    fin_estimulo = inicio_pulso + duracion_pulso

    indices = t > fin_estimulo

    t_post = t[indices]
    v_post = v[indices]
    w_post = w[indices]
    dv_post = dv[indices]

    duracion, t_subida, t_bajada = duracion_por_cruces(t_post, v_post, umbral)

    resultados = {
        "v_max": np.max(v_post),
        "v_min": np.min(v_post),
        "dv_max": np.max(dv_post),
        "dv_min": np.min(dv_post),
        "duracion": duracion,
        "t_subida": t_subida,
        "t_bajada": t_bajada
    }

    return resultados
```

```python
resultados = medir_potencial_de_accion(t, v, w, dv, umbral=-20.0)

for clave, valor in resultados.items():
    print(f"{clave}: {valor:.4f}")
```

$$
\dot v=J_F-g_US_m(v)(v-v_U)-g_Dw(v-v_D),

la Jacobiana en el orden de variables $(w,v)$:

J(w,v)

$$

Por tanto,

$$
\rho'(v)=s_wb_w\alpha(v)+s_w(b_w-1)\beta(v).
$$

El objetivo de esta sección es verificar con `sympy` que la Jacobiana simbólica coincide con la expresión calculada a mano.

#### Definición simbólica del modelo

Definimos las variables dinámicas $w$ y $v$, los parámetros del modelo, la función de activación $S_m(v)$ y las tasas $\alpha(v)$, $\beta(v)$.

```python
import sympy as sp

sp.init_printing()

# Variables dinámicas
w, v = sp.symbols("w v")

# Parámetros
J_F, g_U, g_D = sp.symbols("J_F g_U g_D")
v_U, v_D = sp.symbols("v_U v_D")
v_m, v_w = sp.symbols("v_m v_w")
s_m, s_w, b_w, r = sp.symbols("s_m s_w b_w r")

# Función de activación de la corriente Up
S_m = 1 / (1 + sp.exp(s_m * (v_m - v)))

# Tasas de activación y desactivación de w
alpha = r * sp.exp(s_w * b_w * (v - v_w))
beta = r * sp.exp(s_w * (b_w - 1) * (v - v_w))

# Suma de tasas
rho = alpha + beta

# Sistema en el orden (w, v)
F = alpha * (1 - w) - beta * w
G = J_F - g_U * S_m * (v - v_U) - g_D * w * (v - v_D)

campo = sp.Matrix([F, G])
variables = sp.Matrix([w, v])

# Jacobiana calculada automáticamente
J_sympy = sp.simplify(campo.jacobian(variables))

J_sympy
```

La matriz anterior puede verse complicada porque `sympy` expande las expresiones exponenciales. Para verificar que coincide con nuestra derivación, construimos ahora la matriz obtenida a mano y comparamos ambas.

```python
# Derivadas auxiliares
dS_m = sp.diff(S_m, v)
dalpha = sp.diff(alpha, v)
dbeta = sp.diff(beta, v)
drho = sp.diff(rho, v)

# Jacobiana obtenida a mano
J_manual = sp.Matrix([
    [
        -rho,
        dalpha - w * drho
    ],
    [
        -g_D * (v - v_D),
        -g_U * (dS_m * (v - v_U) + S_m) - g_D * w
    ]
])

J_manual_simplificada = sp.simplify(J_manual)

J_manual_simplificada
```

J_{\text{sympy}}-J_{\text{manual}}.

Si el resultado es la matriz cero, entonces la salida de `sympy` coincide exactamente con el cálculo a mano.

```python
verificacion_J = sp.simplify(J_sympy - J_manual_simplificada)

verificacion_J
```

$$
\dot w=0,
$$

$$
\dot v=0.

$$

Expandiendo,

$$
0=\alpha(v^*)-\alpha(v^*)w^*-\beta(v^*)w^*.
$$

Por tanto,

$$
0=\alpha(v^*)-[\alpha(v^*)+\beta(v^*)]w^*.
$$

Como

$$
\rho(v)=\alpha(v)+\beta(v),
$$

obtenemos

$$
w^*=\frac{\alpha(v^*)}{\rho(v^*)}.
$$

Sustituimos esta condición en la Jacobiana.

```python
# Condición de punto fijo para w
w_estrella = alpha / rho

# Jacobiana evaluada en w*=alpha/rho
J_estrella_sympy = sp.simplify(J_manual_simplificada.subs(w, w_estrella))

J_estrella_sympy
```

La salida sigue viéndose expandida en términos de exponenciales. Queremos verificar que coincide con la expresión compacta obtenida a mano:

\begin{pmatrix}

&
\dfrac{s_w\alpha^*\beta^*}{\rho^*}
\\[10pt]
-g_D(v^*-v_D)

-g_US_m(v^*)
\left[
1+s_m(1-S_m(v^*))(v^*-v_U)
\right]
-g_Dw^*
\end{pmatrix}.
$$

En la verificación simbólica usamos las funciones sin asterisco, entendiendo que todo está evaluado en $v=v^*$.

La salida es nuevamente


### 5. Calcule las expresiones para el conjunto de puntos de inflexión $I$, y la curva de umbrales $I_\theta$.

El objetivo de este punto es encontrar los puntos del plano fase donde la trayectoria temporal $v(t)$ cambia de concavidad.

Es decir, queremos calcular el conjunto $I=\{(w,v):\ddot v=0\}.$

Después, queremos identificar la parte de ese conjunto que corresponde al umbral del potencial de acción. Esta curva se denota por $I_\theta.$

La idea geométrica es que un potencial de acción tiene varios cambios de concavidad, pero el umbral se asocia con el primer punto de inflexión durante la subida rápida del voltaje.

Por eso, la curva de umbrales no será todo $I$, sino solamente la parte de $I$ que ocurre durante la fase inicial de subida del potencial de acción.

Partimos del modelo minimal Up-Down:

$$
\dot w=\alpha(v)-w\rho(v),
$$

$$
\dot v=J_F-g_US_m(v)(v-v_U)-g_Dw(v-v_D),
$$

donde

$$
\rho(v)=\alpha(v)+\beta(v).
$$

Recordemos que

$$
S_m(v)=\frac{1}{1+\exp[s_m(v_m-v)]},
$$

$$
\alpha(v)=r\exp[s_wb_w(v-v_w)],
$$

$$
\beta(v)=r\exp[s_w(b_w-1)(v-v_w)].
$$

Para simplificar la notación, escribimos la ecuación de $v$ como

$$
\dot v=H(v)-wK(v),
$$

donde

$$
H(v)=J_F-g_US_m(v)(v-v_U),
$$

y

$$
K(v)=g_D(v-v_D).
$$

Así, el sistema queda escrito como

$$
\boxed{
\dot v=H(v)-wK(v)
}
$$

$$
\boxed{
\dot w=\alpha(v)-w\rho(v).
}
$$


El conjunto de puntos de inflexión $I$ se obtiene imponiendo

$$
\ddot v=0.
$$

La cantidad $\ddot v$ mide la concavidad de la gráfica temporal $t\mapsto v(t)$:

- Si $\ddot v>0$, la gráfica de $v(t)$ es cóncava hacia arriba.
- Si $\ddot v<0$, la gráfica de $v(t)$ es cóncava hacia abajo.
- Si $\ddot v=0$, hay un punto de inflexión.

Durante la subida de un potencial de acción, el voltaje primero acelera hacia arriba y después comienza a desacelerarse antes de alcanzar su máximo. El punto donde ocurre ese cambio se puede interpretar como un umbral geométrico.

Por eso, una vez obtenido $I$, la curva de umbrales $I_\theta$ será la parte de $I$ que está en la región donde el potencial está subiendo.

Partimos de

$$
\dot v=H(v)-wK(v).
$$

Derivamos con respecto al tiempo:

$$
\ddot v
=
\frac{d}{dt}\left[H(v)-wK(v)\right].
$$

Aplicando la regla de la cadena,

$$
\frac{d}{dt}H(v)=H'(v)\dot v.
$$

Para el segundo término usamos la regla del producto:

$$
\frac{d}{dt}\left[wK(v)\right]
=
\dot wK(v)+wK'(v)\dot v.
$$

Por tanto,

$$
\ddot v
=
H'(v)\dot v-\left[\dot wK(v)+wK'(v)\dot v\right].
$$

Distribuyendo el signo negativo,

$$
\ddot v
=
H'(v)\dot v-\dot wK(v)-wK'(v)\dot v.
$$

Agrupando los términos que contienen $\dot v$,

$$
\ddot v
=
\dot v\left[H'(v)-wK'(v)\right]
-
K(v)\dot w.
$$

Ahora sustituimos

$$
\dot v=H(v)-wK(v),
$$

y

$$
\dot w=\alpha(v)-w\rho(v).
$$

Entonces

$$
\boxed{
\ddot v
=
\left[H(v)-wK(v)\right]
\left[H'(v)-wK'(v)\right]
-
K(v)\left[\alpha(v)-w\rho(v)\right].
}
$$

Esta es la expresión fundamental para calcular el conjunto de puntos de inflexión.


Queremos escribir $\ddot v$ como una expresión cuadrática en $w$.

Partimos de

$$
\ddot v
=
(H-wK)(H'-wK')-K(\alpha-w\rho).
$$

Primero expandimos el producto:

$$
(H-wK)(H'-wK')
=
HH'-HwK'-wKH'+w^2KK'.
$$

Ahora expandimos el término restante:

$$
-K(\alpha-w\rho)
=
-K\alpha+Kw\rho.
$$

Por tanto,

$$
\ddot v
=
HH'-HwK'-wKH'+w^2KK'-K\alpha+Kw\rho.
$$

Agrupamos por potencias de $w$.

El término cuadrático es

$$
w^2KK'.
$$

Los términos lineales en $w$ son

$$
-HwK'-wKH'+Kw\rho.
$$

Factorizando $w$,

$$
w[-HK'-KH'+K\rho].
$$

El término independiente es

$$
HH'-K\alpha.
$$

Por tanto,

$$
\boxed{
\ddot v=a_2(v)w^2+a_1(v)w+a_0(v),
}
$$

donde

$$
\boxed{
a_2(v)=K(v)K'(v),
}
$$

$$
\boxed{
a_1(v)=K(v)\rho(v)-K(v)H'(v)-H(v)K'(v),
}
$$

y

$$
\boxed{
a_0(v)=H(v)H'(v)-K(v)\alpha(v).
}
$$

Ahora calculemos explicitamente $H'(v)$ y $K'(v)$

Recordemos que

$$
H(v)=J_F-g_US_m(v)(v-v_U).
$$

Entonces

$$
H'(v)
=
-g_U\frac{d}{dv}\left[S_m(v)(v-v_U)\right].
$$

Por regla del producto,

$$
\frac{d}{dv}\left[S_m(v)(v-v_U)\right]
=
S_m'(v)(v-v_U)+S_m(v).
$$

Por tanto,

$$
H'(v)
=
-g_U\left[S_m'(v)(v-v_U)+S_m(v)\right].
$$

Como

$$
S_m'(v)=s_mS_m(v)(1-S_m(v)),
$$

se obtiene

$$
H'(v)
=
-g_U\left[
s_mS_m(v)(1-S_m(v))(v-v_U)+S_m(v)
\right].
$$

Factorizando $S_m(v)$,

$$
\boxed{
H'(v)
=
-g_US_m(v)
\left[
1+s_m(1-S_m(v))(v-v_U)
\right].
}
$$

Por otro lado,

$$
K(v)=g_D(v-v_D).
$$

Por tanto,

$$
\boxed{
K'(v)=g_D.
}
$$

Para simplificar la notación, definimos

$$
M(v):=
S_m(v)
\left[
1+s_m(1-S_m(v))(v-v_U)
\right].
$$

Entonces

$$
\boxed{
H'(v)=-g_UM(v).
}
$$

Usando

$$
K(v)=g_D(v-v_D),
$$

$$
K'(v)=g_D,
$$

y

$$
H'(v)=-g_UM(v),
$$

obtenemos los coeficientes del cuadrático.

Primero,

$$
a_2(v)=K(v)K'(v).
$$

Sustituyendo,

$$
a_2(v)=g_D(v-v_D)g_D.
$$

Por tanto,

$$
\boxed{
a_2(v)=g_D^2(v-v_D).
}
$$

Ahora,

$$
a_1(v)=K(v)\rho(v)-K(v)H'(v)-H(v)K'(v).
$$

Sustituyendo $H'(v)=-g_UM(v)$ y $K'(v)=g_D$,

$$
a_1(v)
=
K(v)\rho(v)-K(v)[-g_UM(v)]-H(v)g_D.
$$

Entonces

$$
a_1(v)
=
K(v)\rho(v)+g_UK(v)M(v)-g_DH(v).
$$

Como $K(v)=g_D(v-v_D)$,

$$
\boxed{
a_1(v)
=
g_D(v-v_D)\left[\rho(v)+g_UM(v)\right]
-
g_DH(v).
}
$$

Finalmente,

$$
a_0(v)=H(v)H'(v)-K(v)\alpha(v).
$$

Sustituyendo $H'(v)=-g_UM(v)$,

$$
a_0(v)
=
H(v)[-g_UM(v)]-K(v)\alpha(v).
$$

Por tanto,

$$
a_0(v)
=
-g_UH(v)M(v)-K(v)\alpha(v).
$$

Como $K(v)=g_D(v-v_D)$,

$$
\boxed{
a_0(v)
=
-g_UH(v)M(v)
-
g_D(v-v_D)\alpha(v).
}
$$

Así, el conjunto de puntos de inflexión queda determinado por

$$
\boxed{
a_2(v)w^2+a_1(v)w+a_0(v)=0.
}
$$

Así, el conjunto de puntos de inflexión es

$$
I=\{(w,v):\ddot v(w,v)=0\}.
$$

Como

$$
\ddot v=a_2(v)w^2+a_1(v)w+a_0(v),
$$

entonces $I$ está dado por las soluciones de

$$
a_2(v)w^2+a_1(v)w+a_0(v)=0.
$$

Si

$$
a_2(v)\neq 0,
$$

podemos resolver para $w$ usando la fórmula general de la ecuación cuadrática:

$$
w=
\frac{
-a_1(v)\pm\sqrt{a_1(v)^2-4a_2(v)a_0(v)}
}{
2a_2(v)
}.
$$

Definimos

$$
\boxed{
D_I(v)=a_1(v)^2-4a_2(v)a_0(v).
}
$$

Entonces las dos ramas del conjunto de inflexión son

$$
\boxed{
w_-(v)
=
\frac{
-a_1(v)-\sqrt{D_I(v)}
}{
2a_2(v)
},
}
$$

y

$$
\boxed{
w_+(v)
=
\frac{
-a_1(v)+\sqrt{D_I(v)}
}{
2a_2(v)
}.
}
$$

Por tanto,

$$
\boxed{
I
=
\left\{
(w,v):
w=w_-(v)
\ \text{o}\
w=w_+(v)
\right\}.
}
$$

Equivalentemente,

$$
\boxed{
I
=
\left\{
(w,v):
w=
\frac{
-a_1(v)\pm\sqrt{D_I(v)}
}{
2a_2(v)
}
\right\}.
}
$$

Sólo tienen sentido real los valores de $v$ para los cuales

$$
D_I(v)\geq 0.
$$

Además, como el dominio físico del modelo es

$$
v\in[v_D,v_U],
$$

$$
w\in[0,1],
$$

tomamos únicamente la parte de $I$ que pertenece al dominio:

$$
\boxed{
I
=
\left\{
(w,v)\in[v_D,v_U]\times[0,1]:
\ddot v(w,v)=0
\right\}.
}
$$


El conjunto $I$ contiene todos los puntos de inflexión, pero no todos corresponden al umbral del potencial de acción.

El umbral se asocia con el primer punto de inflexión durante la subida rápida de $v(t)$. Por eso restringimos a la región donde

$$
\dot v>0
$$

y

$$
\dot w>0.
$$

Esta región la denotamos por $R_1$.

Como

$$
\dot v=H(v)-wK(v),
$$

la condición $\dot v>0$ es

$$
H(v)-wK(v)>0.
$$

Si $K(v)>0$, esto equivale a

$$
w<\frac{H(v)}{K(v)}.
$$

La curva

$$
w=\frac{H(v)}{K(v)}
$$

es la ceroclina de $v$.

Por otro lado,

$$
\dot w=\alpha(v)-w\rho(v).
$$

Como

$$
\rho(v)>0,
$$

la condición $\dot w>0$ equivale a

$$
w<\frac{\alpha(v)}{\rho(v)}.
$$

La curva

$$
w=\frac{\alpha(v)}{\rho(v)}
$$

es la ceroclina de $w$.

Por tanto,

$$
\boxed{
R_1
=
\left\{
(w,v):
0\leq w\leq 1,\ 
v_D<v<v_U,\ 
w<\frac{H(v)}{K(v)},\ 
w<\frac{\alpha(v)}{\rho(v)}
\right\}.
}
$$

De manera compacta,

$$
\boxed{
R_1
=
\left\{
(w,v):
0\leq w<\min\left\{
\frac{H(v)}{K(v)},\frac{\alpha(v)}{\rho(v)}
\right\}
\right\}.
}
$$

La curva de umbrales se define como la parte del conjunto de inflexión que cae dentro de la región de subida inicial:

$$
\boxed{
I_\theta=I\cap R_1.
}
$$

Es decir,

$$
\boxed{
I_\theta
=
\left\{
(w,v):
\ddot v(w,v)=0,\ 
\dot v(w,v)>0,\ 
\dot w(w,v)>0
\right\}.
}
$$

Usando las ramas $w_-(v)$ y $w_+(v)$, podemos escribir

$$
\boxed{
I_\theta
=
\left\{
(w_\pm(v),v):
D_I(v)\geq 0,\ 
0\leq w_\pm(v)\leq 1,\ 
w_\pm(v)<\frac{H(v)}{K(v)},\ 
w_\pm(v)<\frac{\alpha(v)}{\rho(v)}
\right\}.
}
$$

En la práctica, la rama que suele corresponder al primer punto de inflexión durante la subida del potencial de acción es la rama inferior,

$$
w_-(v).
$$

Por eso, una forma útil de escribir la curva de umbrales es

$$
\boxed{
I_\theta
\approx
\left\{
(w_-(v),v):
D_I(v)\geq 0,\ 
0\leq w_-(v)\leq 1,\ 
w_-(v)<\min\left\{
\frac{H(v)}{K(v)},\frac{\alpha(v)}{\rho(v)}
\right\}
\right\}.
}
$$


### 6. Encuentre una manera  de medir el área del conjunto cuya frontera es $I_\theta$.

**Pista:** puede intentar una medida que cuenta.


En el punto anterior se calculó el conjunto de puntos de inflexión

$$
I=\{(w,v):\ddot v=0\},
$$

y la curva de umbrales

$$
I_\theta=I\cap R_1.
$$

Ahora queremos medir el área del conjunto cuya frontera está determinada por la curva de umbrales $I_\theta$.


El modelo minimal se estudia en el dominio

$$
D=[v_D,v_U]\times[0,1].
$$

Aquí:

- $v$ representa el potencial transmembranal,
- $w$ representa la variable de recuperación,
- $v_D$ es el potencial bajo o descendente,
- $v_U$ es el potencial alto o ascendente,
- $w\in[0,1]$ representa una proporción de canales abiertos.

Entonces el área total del dominio es

$$
|D|=(v_U-v_D)(1).
$$

Como el intervalo de $w$ tiene longitud $1$, queda

$$
\boxed{
|D|=v_U-v_D.
}
$$

Por ejemplo, si

$$
v_D=-80,
$$

y

$$
v_U=60,
$$

entonces

$$
|D|=60-(-80)=140.
$$


Del punto anterior, la curva de umbrales se obtuvo como

$$
I_\theta=I\cap R_1.
$$

El conjunto $I$ está determinado por

$$
\ddot v=0.
$$

Como se obtuvo que

$$
\ddot v=a_2(v)w^2+a_1(v)w+a_0(v),
$$

las ramas de $I$ son

$$
w_\pm(v)
=
\frac{
-a_1(v)\pm\sqrt{D_I(v)}
}{
2a_2(v)
},
$$

donde

$$
D_I(v)=a_1(v)^2-4a_2(v)a_0(v).
$$

La curva de umbrales $I_\theta$ corresponde a la parte de estas ramas que cae dentro de la región $R_1$, es decir, dentro de la región donde

$$
\dot v>0,
$$

y

$$
\dot w>0.
$$

En términos de las ceroclinas, esto significa

$$
w<\frac{H(v)}{K(v)},
$$

y

$$
w<\frac{\alpha(v)}{\rho(v)}.
$$

Por tanto,

$$
R_1
=
\left\{
(w,v):
w<\min\left\{
\frac{H(v)}{K(v)},\frac{\alpha(v)}{\rho(v)}
\right\}
\right\}.
$$

Así,

$$
I_\theta
=
\left\{
(w_\pm(v),v):
D_I(v)\geq0,\ 
0\leq w_\pm(v)\leq1,\ 
w_\pm(v)<\min\left\{
\frac{H(v)}{K(v)},\frac{\alpha(v)}{\rho(v)}
\right\}
\right\}.
$$

En la práctica, la rama que suele asociarse al primer punto de inflexión durante la subida del potencial de acción es la rama inferior,

$$
w_-(v).
$$

Por eso podemos escribir, de forma práctica,

$$
w_\theta(v)=w_-(v),
$$

siempre que dicha rama pertenezca a $R_1$.


Una vez seleccionada la curva de umbrales, podemos definir el conjunto cuya frontera está dada por $I_\theta$ como la región bajo la curva de umbrales:

$$
A_\theta
=
\{(v,w):v\in V_\theta,\ 0\leq w\leq w_\theta(v)\}.
$$

Aquí $V_\theta$ es el conjunto de valores de $v$ para los cuales la curva de umbrales está bien definida.

Es decir,

$$
v\in V_\theta
$$

si se cumplen las condiciones:

$$
D_I(v)\geq0,
$$

$$
0\leq w_\theta(v)\leq1,
$$

y

$$
w_\theta(v)<\min\left\{
\frac{H(v)}{K(v)},\frac{\alpha(v)}{\rho(v)}
\right\}.
$$

Entonces el conjunto $A_\theta$ se puede escribir como

$$
\boxed{
A_\theta
=
\left\{
(v,w):
v\in V_\theta,\ 0\leq w\leq w_\theta(v)
\right\}.
}
$$

Esta región representa el conjunto de puntos del plano fase que quedan debajo de la curva de umbrales.


Si la curva $w_\theta(v)$ está definida en un intervalo continuo $[a,b]$, entonces el área bajo la curva se puede calcular mediante la integral

$$
|A_\theta|
=
\int_a^b w_\theta(v)\,dv.
$$

Más generalmente, si $V_\theta$ no es un intervalo único, sino una unión de intervalos,

$$
V_\theta=\bigcup_{k=1}^n [a_k,b_k],
$$

entonces el área se puede calcular como

$$
\boxed{
|A_\theta|
=
\sum_{k=1}^n
\int_{a_k}^{b_k}w_\theta(v)\,dv.
}
$$

Esta es una forma matemática válida de medir el área.

Sin embargo, en la práctica la curva $I_\theta$ puede estar definida solamente en ciertos intervalos, puede tener huecos o puede ser difícil de integrar analíticamente. Por eso usaremos una medida que cuenta.

Una forma práctica de medir el área es discretizar el dominio del plano fase y contar cuántos puntos o celdas pertenecen al conjunto $A_\theta$.

Tomamos una malla rectangular sobre

$$
D=[v_D,v_U]\times[0,1].
$$

Dividimos el intervalo de $v$ en $N_v$ partes y el intervalo de $w$ en $N_w$ partes.

Entonces

$$
\Delta v=\frac{v_U-v_D}{N_v},
$$

y

$$
\Delta w=\frac{1}{N_w}.
$$

Cada celda de la malla tiene área

$$
\Delta A=\Delta v\,\Delta w.
$$

Ahora tomamos puntos de la malla

$$
(v_i,w_j),
$$

y definimos la función indicadora del conjunto $A_\theta$:

$$
\mathbf{1}_{A_\theta}(v_i,w_j)
=
\begin{cases}
1, & \text{si } (v_i,w_j)\in A_\theta,\\
0, & \text{si } (v_i,w_j)\notin A_\theta.
\end{cases}
$$

Entonces el área se aproxima mediante

$$
\boxed{
|A_\theta|
\approx
\Delta v\,\Delta w
\sum_{i=1}^{N_v}
\sum_{j=1}^{N_w}
\mathbf{1}_{A_\theta}(v_i,w_j).
}
$$

Esta fórmula es la medida que cuenta: contamos cuántas celdas de la malla caen dentro de la región y multiplicamos por el área de cada celda.

Para decidir si un punto de la malla pertenece a $A_\theta$, usamos la definición

$$
A_\theta
=
\{(v,w):v\in V_\theta,\ 0\leq w\leq w_\theta(v)\}.
$$

Entonces

$$
(v_i,w_j)\in A_\theta
$$

si y sólo si se cumplen las condiciones:

$$
v_i\in V_\theta,
$$

y

$$
0\leq w_j\leq w_\theta(v_i).
$$

A su vez, $v_i\in V_\theta$ significa que:

$$
D_I(v_i)\geq0,
$$

$$
0\leq w_\theta(v_i)\leq1,
$$

y

$$
w_\theta(v_i)<\min\left\{
\frac{H(v_i)}{K(v_i)},
\frac{\alpha(v_i)}{\rho(v_i)}
\right\}.
$$

Por tanto, la función indicadora se puede escribir como

$$
\mathbf{1}_{A_\theta}(v_i,w_j)
=
\mathbf{1}
\left[
D_I(v_i)\geq0
\right]
\mathbf{1}
\left[
0\leq w_\theta(v_i)\leq1
\right]
\mathbf{1}
\left[
w_\theta(v_i)<\min\left\{
\frac{H(v_i)}{K(v_i)},
\frac{\alpha(v_i)}{\rho(v_i)}
\right\}
\right]
\mathbf{1}
\left[
0\leq w_j\leq w_\theta(v_i)
\right].
$$

Esto permite calcular el área mediante una suma finita.


Como el dominio total tiene área

$$
|D|=v_U-v_D,
$$

podemos normalizar el área de $A_\theta$ dividiendo entre $|D|$:

$$
P_\theta
=
\frac{|A_\theta|}{|D|}.
$$

Entonces

$$
\boxed{
P_\theta
=
\frac{|A_\theta|}{v_U-v_D}.
}
$$

Esta cantidad puede interpretarse como una probabilidad geométrica: si se escoge una condición inicial al azar de manera uniforme en el dominio $D$, entonces $P_\theta$ aproxima la probabilidad de que esa condición inicial caiga dentro del conjunto $A_\theta$.

En términos discretos,

$$
P_\theta
\approx
\frac{
\Delta v\,\Delta w
\sum_{i=1}^{N_v}
\sum_{j=1}^{N_w}
\mathbf{1}_{A_\theta}(v_i,w_j)
}{
(v_U-v_D)
}.
$$

Como

$$
\Delta v\,\Delta w
=
\frac{v_U-v_D}{N_v}
\frac{1}{N_w},
$$

se obtiene

$$
P_\theta
\approx
\frac{
\frac{v_U-v_D}{N_vN_w}
\sum_{i=1}^{N_v}
\sum_{j=1}^{N_w}
\mathbf{1}_{A_\theta}(v_i,w_j)
}{
v_U-v_D
}.
$$

Cancelando,

$$
\boxed{
P_\theta
\approx
\frac{1}{N_vN_w}
\sum_{i=1}^{N_v}
\sum_{j=1}^{N_w}
\mathbf{1}_{A_\theta}(v_i,w_j).
}
$$

Es decir, la fracción normalizada del área es simplemente la proporción de puntos de la malla que caen dentro del conjunto.


La curva $I_\theta$ separa una región del plano fase relacionada con el umbral de excitabilidad.

Medir el área del conjunto delimitado por $I_\theta$ nos da una forma de cuantificar el tamaño de dicha región.

Si el área $|A_\theta|$ aumenta, significa que la región asociada con el umbral ocupa una porción más grande del plano fase. Si disminuye, significa que dicha región se vuelve más pequeña.

Esto puede usarse para estudiar cómo cambia la excitabilidad del sistema al variar parámetros. Por ejemplo:

- Si aumenta la corriente ascendente, la región asociada al disparo puede crecer.
- Si aumenta la corriente descendente, la región asociada al disparo puede reducirse.
- Si la corriente ascendente se activa a voltajes más bajos, puede aumentar la región excitable.
- Si la recuperación se activa demasiado pronto, puede reducirse la región excitable.

Así, el área bajo $I_\theta$ puede funcionar como una medida geométrica de excitabilidad.


La medida por conteo tiene varias ventajas:

1. No requiere una fórmula cerrada simple para el área.

2. Funciona incluso si $I_\theta$ está definida en varios intervalos separados.

3. Permite aproximar áreas de regiones complicadas del plano fase.

4. Es fácil de implementar computacionalmente.

5. Se conecta de manera natural con una interpretación probabilística.

Por eso, aunque matemáticamente podríamos escribir

$$
|A_\theta|=\int_{V_\theta} w_\theta(v)\,dv,
$$

en la práctica es más robusto usar

$$
|A_\theta|
\approx
\Delta v\,\Delta w
\sum_{i=1}^{N_v}
\sum_{j=1}^{N_w}
\mathbf{1}_{A_\theta}(v_i,w_j).
$$


Una manera matemática y práctica de medir el área del conjunto cuya frontera es $I_\theta$ es definir la región

$$
A_\theta
=
\left\{
(v,w):
v\in V_\theta,\ 0\leq w\leq w_\theta(v)
\right\},
$$

donde $w_\theta(v)$ es la rama de $I_\theta$ que corresponde al primer punto de inflexión durante la subida del potencial de acción.

El área exacta se puede escribir como

$$
|A_\theta|
=
\int_{V_\theta}w_\theta(v)\,dv.
$$

Si $V_\theta$ es una unión de intervalos,

$$
V_\theta=\bigcup_{k=1}^n[a_k,b_k],
$$

entonces

$$
|A_\theta|
=
\sum_{k=1}^n
\int_{a_k}^{b_k}w_\theta(v)\,dv.
$$

Sin embargo, una forma práctica, inspirada en la pista de usar una medida que cuenta, consiste en discretizar el dominio

$$
D=[v_D,v_U]\times[0,1],
$$

y aproximar el área mediante

$$
\boxed{
|A_\theta|
\approx
\Delta v\,\Delta w
\sum_{i=1}^{N_v}
\sum_{j=1}^{N_w}
\mathbf{1}_{A_\theta}(v_i,w_j).
}
$$

Aquí

$$
\Delta v=\frac{v_U-v_D}{N_v},
$$

$$
\Delta w=\frac{1}{N_w},
$$

y

$$
\mathbf{1}_{A_\theta}(v_i,w_j)
=
\begin{cases}
1, & \text{si } (v_i,w_j)\in A_\theta,\\
0, & \text{en otro caso}.
\end{cases}
$$

La versión normalizada del área es

$$
\boxed{
P_\theta
\approx
\frac{1}{N_vN_w}
\sum_{i=1}^{N_v}
\sum_{j=1}^{N_w}
\mathbf{1}_{A_\theta}(v_i,w_j).
}
$$

Esta cantidad representa la fracción del plano fase ocupada por el conjunto delimitado por la curva de umbrales.

```python
# Parámetros del modelo minimal Up-Down

parametros = {
    "gU": 2.0,
    "gD": 10.0,
    "r": 0.1,
    "vU": 60.0,
    "vD": -80.0,
    "vm": -20.0,
    "vw": 0.0,
    "sm": 0.22,
    "sw": 0.11,
    "bw": 0.35,
    "JF": 0.0
}
```

```python
def construir_funciones_area(parametros):
    """
    Construye las funciones necesarias para calcular I_theta
    y aproximar el área del conjunto delimitado por la curva de umbrales.
    """

    gU = parametros["gU"]
    gD = parametros["gD"]
    r = parametros["r"]
    vU = parametros["vU"]
    vD = parametros["vD"]
    vm = parametros["vm"]
    vw = parametros["vw"]
    sm = parametros["sm"]
    sw = parametros["sw"]
    bw = parametros["bw"]
    JF = parametros["JF"]

    def Sm(v):
        return 1.0 / (1.0 + np.exp(sm * (vm - v)))

    def alpha(v):
        return r * np.exp(sw * bw * (v - vw))

    def beta(v):
        return r * np.exp(sw * (bw - 1.0) * (v - vw))

    def rho(v):
        return alpha(v) + beta(v)

    def Sw(v):
        return alpha(v) / rho(v)

    def H(v):
        return JF - gU * Sm(v) * (v - vU)

    def K(v):
        return gD * (v - vD)

    def Nv(v):
        return H(v) / K(v)

    def M(v):
        return Sm(v) * (1.0 + sm * (1.0 - Sm(v)) * (v - vU))

    def Hprima(v):
        return -gU * M(v)

    def Kprima(v):
        return gD + 0.0 * v

    def a2(v):
        return K(v) * Kprima(v)

    def a1(v):
        return K(v) * rho(v) - K(v) * Hprima(v) - H(v) * Kprima(v)

    def a0(v):
        return H(v) * Hprima(v) - K(v) * alpha(v)

    def discriminante_inflexion(v):
        return a1(v)**2 - 4.0 * a2(v) * a0(v)

    def rama_inferior(v):
        """
        Rama w_-(v) del conjunto de inflexión.
        """
        D = discriminante_inflexion(v)

        w_theta = np.full_like(v, np.nan, dtype=float)

        mascara = D >= 0.0

        w_theta[mascara] = (
            -a1(v[mascara]) - np.sqrt(D[mascara])
        ) / (2.0 * a2(v[mascara]))

        return w_theta

    def rama_superior(v):
        """
        Rama w_+(v) del conjunto de inflexión.
        """
        D = discriminante_inflexion(v)

        w_theta = np.full_like(v, np.nan, dtype=float)

        mascara = D >= 0.0

        w_theta[mascara] = (
            -a1(v[mascara]) + np.sqrt(D[mascara])
        ) / (2.0 * a2(v[mascara]))

        return w_theta

    return {
        "Sm": Sm,
        "alpha": alpha,
        "beta": beta,
        "rho": rho,
        "Sw": Sw,
        "H": H,
        "K": K,
        "Nv": Nv,
        "a2": a2,
        "a1": a1,
        "a0": a0,
        "discriminante_inflexion": discriminante_inflexion,
        "rama_inferior": rama_inferior,
        "rama_superior": rama_superior
    }
```

```python
def medir_area_Itheta_contando(
    parametros,
    n_v=2000,
    n_w=1000,
    rama="inferior",
    lado="debajo"
):
    """
    Aproxima el área del conjunto delimitado por I_theta usando conteo en una malla.

    Parámetros
    -
    parametros : dict
        Diccionario con los parámetros del modelo.

    n_v : int
        Número de divisiones en la dirección de v.

    n_w : int
        Número de divisiones en la dirección de w.

    rama : str
        Puede ser "inferior" o "superior".
        Para el umbral usualmente usamos la rama "inferior".

    lado : str
        Puede ser "debajo" o "encima".
        Para medir el área bajo la curva usamos "debajo".

    Retorna
    -
    resultados : dict
        Diccionario con el área aproximada, fracción normalizada,
        malla y curva de umbral.
    """

    funcs = construir_funciones_area(parametros)

    vD = parametros["vD"]
    vU = parametros["vU"]

    # Tamaños de celda
    delta_v = (vU - vD) / n_v
    delta_w = 1.0 / n_w

    # Centros de celda
    valores_v = np.linspace(vD + delta_v / 2.0, vU - delta_v / 2.0, n_v)
    valores_w = np.linspace(delta_w / 2.0, 1.0 - delta_w / 2.0, n_w)

    # Elegimos la rama de I
    if rama == "inferior":
        w_theta = funcs["rama_inferior"](valores_v)
    elif rama == "superior":
        w_theta = funcs["rama_superior"](valores_v)
    else:
        raise ValueError("rama debe ser 'inferior' o 'superior'.")

    # Condiciones para que la curva pertenezca a I_theta = I intersectado R1
    D_I = funcs["discriminante_inflexion"](valores_v)
    Nv = funcs["Nv"](valores_v)
    Sw = funcs["Sw"](valores_v)

    curva_valida = (
        (D_I >= 0.0)
        & np.isfinite(w_theta)
        & (w_theta >= 0.0)
        & (w_theta <= 1.0)
        & (w_theta < Nv)
        & (w_theta < Sw)
    )

    # Malla en w
    W = valores_w[None, :]

    if lado == "debajo":
        puntos_dentro = curva_valida[:, None] & (W <= w_theta[:, None])
    elif lado == "encima":
        puntos_dentro = curva_valida[:, None] & (W >= w_theta[:, None])
    else:
        raise ValueError("lado debe ser 'debajo' o 'encima'.")

    # Área aproximada por conteo de celdas
    numero_celdas = np.sum(puntos_dentro)

    area = numero_celdas * delta_v * delta_w

    area_total = (vU - vD) * 1.0
    fraccion_normalizada = area / area_total

    resultados = {
        "area": area,
        "fraccion_normalizada": fraccion_normalizada,
        "numero_celdas": numero_celdas,
        "delta_v": delta_v,
        "delta_w": delta_w,
        "valores_v": valores_v,
        "valores_w": valores_w,
        "w_theta": w_theta,
        "curva_valida": curva_valida,
        "Nv": Nv,
        "Sw": Sw,
        "puntos_dentro": puntos_dentro
    }

    return resultados
```

```python
resultados_area = medir_area_Itheta_contando(
    parametros,
    n_v=5000,
    n_w=2000,
    rama="inferior",
    lado="debajo"
)

print("Área aproximada:", resultados_area["area"])
print("Fracción normalizada:", resultados_area["fraccion_normalizada"])
print("Número de celdas contadas:", resultados_area["numero_celdas"])
```

```python
def graficar_area_Itheta_con_streamplot(resultados_area, parametros, densidad=3.2):
    """
    Grafica el plano fase con ejes invertidos:
    
        eje horizontal: w
        eje vertical: v
    
    Incluye:
    - streamplot del campo vectorial,
    - ceroclinas,
    - curva de umbral I_theta,
    - región cuya área se cuenta.
    """

    valores_v = resultados_area["valores_v"]
    w_theta = resultados_area["w_theta"]
    curva_valida = resultados_area["curva_valida"]
    Nv = resultados_area["Nv"]
    Sw = resultados_area["Sw"]

    # Reconstruimos funciones del modelo
    funcs = construir_funciones_area(parametros)

    H = funcs["H"]
    K = funcs["K"]
    alpha = funcs["alpha"]
    rho = funcs["rho"]

    vD = parametros["vD"]
    vU = parametros["vU"]

    # Malla para el campo vectorial
    valores_w_stream = np.linspace(0, 1, 80)
    valores_v_stream = np.linspace(vD, vU, 80)

    W, V = np.meshgrid(valores_w_stream, valores_v_stream)

    # Campo vectorial en coordenadas (w, v)
    dW = alpha(V) - W * rho(V)
    dV = H(V) - W * K(V)

    plt.figure(figsize=(7, 8))

    # Campo vectorial
    plt.streamplot(
        W,
        V,
        dW,
        dV,
        density=densidad,
        linewidth=0.8,
        arrowsize=1.0,
        color="0.65"
    )

    # Ceroclina de v: w = Nv(v)
    plt.plot(
        Nv,
        valores_v,
        linewidth=2,
        label=r"$N_v:\ \dot v=0$"
    )

    # Ceroclina de w: w = Sw(v)
    plt.plot(
        Sw,
        valores_v,
        linewidth=2,
        label=r"$N_w:\ \dot w=0$"
    )

    # Región bajo la curva de umbral.
    # Como ahora v es eje vertical, usamos fill_betweenx.
    plt.fill_betweenx(
        valores_v[curva_valida],
        0,
        w_theta[curva_valida],
        alpha=0.3,
        label=r"Área contada"
    )

    # Curva de umbral I_theta
    plt.plot(
        w_theta[curva_valida],
        valores_v[curva_valida],
        color="black",
        linewidth=2.5,
        label=r"$I_\theta$"
    )

    plt.xlabel(r"$w$")
    plt.ylabel(r"$v$ (mV)")
    plt.title(r"Área delimitada por $I_\theta$ en el plano fase $(w,v)$")

    plt.xlim(0, 1)
    plt.ylim(vD, vU)

    plt.grid(True)
    plt.legend(loc="best")
    plt.show()
```



#### 7. Obtenga una expresión para puntos fijos con la que se pueda obtener $J_F$. Use su expresión para obtener un vector de valores $J_F$ dado un vector de valores

$$
v^*=[v_1,\dots,v_n].
$$

También obtenga los valores de $w^*$. ¿Cómo? Refine su vector $v^*$ y repita.


Un punto fijo es un estado donde las dos variables dejan de cambiar:

$$
\dot v=0,
\qquad
\dot w=0.
$$

La estrategia será usar $v^*$ como parámetro. Es decir, en lugar de fijar primero $J_F$ y buscar puntos fijos, haremos lo contrario:

$$
v^*
\longrightarrow
w^*
\longrightarrow
J_F.
$$

Esto nos permite construir una curva de equilibrio de la forma

$$
(J_F(v^*),v^*).
$$

Esta curva será fundamental para estudiar bifurcaciones, porque nos permite ver cuántos puntos fijos existen para cada valor de $J_F$.



De la condición $\dot w=0$, se obtiene que el valor estacionario de $w$ debe satisfacer

$$
w^*=S_w(v^*).
$$

Por tanto, una vez elegido un valor de $v^*$, el valor correspondiente de $w^*$ se obtiene directamente evaluando la función estacionaria $S_w$:

$$
w^*(v^*)=S_w(v^*).
$$

Esta es la primera parte de la parametrización de los puntos fijos.


Ahora usamos la condición $\dot v=0$. Al despejar $J_F$, obtenemos

$$
J_F(v^*)
=
g_US_m(v^*)(v^*-v_U)
+
g_DS_w(v^*)(v^*-v_D).
$$

Esta expresión responde la pregunta central del punto: para cada posible valor de equilibrio $v^*$, nos dice qué corriente externa $J_F$ se necesita para sostener ese equilibrio.

La parametrización de la rama de puntos fijos queda entonces dada por

$$
v^*
\mapsto
\left(
J_F(v^*),w^*(v^*),v^*
\right).
$$

El intervalo natural para tomar los valores de $v^*$ es el dominio fisiológico del modelo:

$$
v^*\in[v_D,v_U].
$$

En las simulaciones usaremos un vector de valores en ese intervalo.

```python
parametros = {
    "gU": 2.0,
    "gD": 10.0,
    "r": 0.1,
    "vU": 60.0,
    "vD": -80.0,
    "vm": -20.0,
    "vw": 0.0,
    "sm": 0.22,
    "sw": 0.11,
    "bw": 0.35,
    "JF": 0.0
}
```



Definimos las funciones de activación y la función $J_F(v^*)$.

La función `JF_equilibrio(v)` implementa la expresión

$$
J_F(v^*)
=
g_US_m(v^*)(v^*-v_U)
+
g_DS_w(v^*)(v^*-v_D).
$$

La función `Sw(v)` nos da directamente los valores estacionarios

$$
w^*=S_w(v^*).
$$

```python
def construir_funciones_puntos_fijos(parametros):
    """
    Construye las funciones necesarias para parametrizar la curva de puntos fijos.
    """

    gU = parametros["gU"]
    gD = parametros["gD"]
    r = parametros["r"]
    vU = parametros["vU"]
    vD = parametros["vD"]
    vm = parametros["vm"]
    vw = parametros["vw"]
    sm = parametros["sm"]
    sw = parametros["sw"]
    bw = parametros["bw"]

    def Sm(v):
        return 1.0 / (1.0 + np.exp(sm * (vm - v)))

    def alpha(v):
        return r * np.exp(sw * bw * (v - vw))

    def beta(v):
        return r * np.exp(sw * (bw - 1.0) * (v - vw))

    def rho(v):
        return alpha(v) + beta(v)

    def Sw(v):
        return alpha(v) / rho(v)

    def JF_equilibrio(v):
        return gU * Sm(v) * (v - vU) + gD * Sw(v) * (v - vD)

    return Sm, alpha, beta, rho, Sw, JF_equilibrio
```

#### 8. Aproxime el valor de $v^*$ para $J_F^*$.

En el punto anterior construimos la función

$$
J_F(v^*)
=
g_US_m(v^*)(v^*-v_U)
+
g_DS_w(v^*)(v^*-v_D).
$$

Ahora queremos resolver el problema inverso:

$$
J_F(v^*)=J_F^*.
$$

Es decir, dado un valor fijo de la corriente externa $J_F^*$, queremos encontrar los valores de equilibrio $v^*$ que corresponden a esa corriente.

Definimos

$$
\Phi(v)=J_F(v)-J_F^*.
$$

Entonces buscamos las raíces de

$$
\Phi(v^*)=0.
$$

Como la curva $J_F(v^*)$ puede no ser monótona, puede haber más de una solución.

**Interpretación geométrica**

En la gráfica $(J_F,v^*)$, fijar un valor $J_F=J_F^*$ equivale a trazar una línea vertical.

Cada intersección entre esa línea vertical y la curva de puntos fijos representa un punto fijo del sistema.

Por tanto:

- una intersección significa que hay un punto fijo;
- tres intersecciones significan que hay tres puntos fijos;
- una tangencia puede indicar una bifurcación tipo saddle-node.

El intervalo natural donde buscamos soluciones es

$$
v^*\in[v_D,v_U].
$$

```python
from scipy.optimize import brentq
```

**Función para encontrar raíces**

La siguiente función busca valores $v^*$ tales que

$$
J_F(v^*)=J_F^*.
$$

Para hacerlo, evalúa

$$
\Phi(v)=J_F(v)-J_F^*
$$

en una malla fina y detecta intervalos donde hay cambios de signo. Después refina cada raíz usando el método de Brent.

```python
def encontrar_v_estrellas_para_JF(
    JF_objetivo,
    JF_equilibrio,
    v_min,
    v_max,
    n_malla=50000,
    tolerancia_duplicados=1e-6
):
    """
    Encuentra valores v* tales que JF_equilibrio(v*) = JF_objetivo.

    Parámetros
    -
    JF_objetivo : float
        Valor de corriente externa JF*.

    JF_equilibrio : callable
        Función JF(v*) obtenida de la condición de punto fijo.

    v_min, v_max : float
        Extremos del intervalo donde se buscan raíces.

    n_malla : int
        Número de puntos para detectar cambios de signo.

    tolerancia_duplicados : float
        Tolerancia para eliminar raíces repetidas.

    Retorna
    -
    np.array
        Arreglo con las raíces v* encontradas.
    """

    valores_v = np.linspace(v_min, v_max, n_malla)
    valores_phi = JF_equilibrio(valores_v) - JF_objetivo

    raices = []

    for i in range(len(valores_v) - 1):
        phi_izq = valores_phi[i]
        phi_der = valores_phi[i + 1]

        if not np.isfinite(phi_izq) or not np.isfinite(phi_der):
            continue

        if abs(phi_izq) < 1e-12:
            raices.append(valores_v[i])

        if phi_izq * phi_der < 0:
            raiz = brentq(
                lambda x: JF_equilibrio(x) - JF_objetivo,
                valores_v[i],
                valores_v[i + 1]
            )
            raices.append(raiz)

    raices_unicas = []

    for raiz in raices:
        if not any(abs(raiz - r) < tolerancia_duplicados for r in raices_unicas):
            raices_unicas.append(raiz)

    return np.array(raices_unicas)
```

**Caso $J_F^*=0$**

Primero aproximamos los puntos fijos cuando no hay corriente externa constante:

$$
J_F^*=0.
$$

Esto significa que buscamos los valores $v^*$ tales que

$$
J_F(v^*)=0.
$$

```python
JF_objetivo = 0.0

vD = parametros["vD"]
vU = parametros["vU"]

v_estrellas_para_JF = encontrar_v_estrellas_para_JF(
    JF_objetivo=JF_objetivo,
    JF_equilibrio=JF_equilibrio,
    v_min=vD,
    v_max=vU,
    n_malla=50000
)

v_estrellas_para_JF
```

**Valores correspondientes de $w^*$**

Una vez encontrados los valores $v^*$, obtenemos los valores correspondientes de $w^*$ usando

$$
w^*=S_w(v^*).
$$

Esto se hace evaluando la función `Sw` en cada raíz encontrada.

#### 9. Clasificación de los puntos fijos

En el punto anterior encontramos los valores $v^*$ que satisfacen

$$
J_F(v^*)=J_F^*.
$$

Ahora evaluamos en esos mismos valores:

$$
\tau(v^*)=\operatorname{tr}(J^*),
$$

$$
\delta(v^*)=\det(J^*),
$$

y

$$
\Delta(v^*)=\tau(v^*)^2-4\delta(v^*).
$$

Con estos tres valores clasificamos cada punto fijo.

```python
import pandas as pd
```

#### Funciones para traza, determinante y discriminante

Usamos las expresiones compactas:

$$
L_m(v)=1+s_m(1-S_m(v))(v-v_U),
$$

$$
\tau(v)
=
-\rho(v)
-
g_US_m(v)L_m(v)
-
g_DS_w(v),
$$

y

$$
\delta(v)
=
\rho(v)
\left(
g_US_m(v)L_m(v)+g_DS_w(v)
\right)
+
g_D(v-v_D)\frac{s_w\alpha(v)\beta(v)}{\rho(v)}.
$$

```python
def construir_funciones_clasificacion(parametros):
    """
    Construye funciones para evaluar traza, determinante,
    discriminante y clasificar puntos fijos.
    """

    gU = parametros["gU"]
    gD = parametros["gD"]
    r = parametros["r"]
    vU = parametros["vU"]
    vD = parametros["vD"]
    vm = parametros["vm"]
    vw = parametros["vw"]
    sm = parametros["sm"]
    sw = parametros["sw"]
    bw = parametros["bw"]

    def Sm(v):
        return 1.0 / (1.0 + np.exp(sm * (vm - v)))

    def alpha(v):
        return r * np.exp(sw * bw * (v - vw))

    def beta(v):
        return r * np.exp(sw * (bw - 1.0) * (v - vw))

    def rho(v):
        return alpha(v) + beta(v)

    def Sw(v):
        return alpha(v) / rho(v)

    def Lm(v):
        return 1.0 + sm * (1.0 - Sm(v)) * (v - vU)

    def traza(v):
        return -rho(v) - gU * Sm(v) * Lm(v) - gD * Sw(v)

    def determinante(v):
        return (
            rho(v) * (gU * Sm(v) * Lm(v) + gD * Sw(v))
            +
            gD * (v - vD) * sw * alpha(v) * beta(v) / rho(v)
        )

    def discriminante(v):
        return traza(v)**2 - 4.0 * determinante(v)

    return Sm, alpha, beta, rho, Sw, Lm, traza, determinante, discriminante
```

```python
Sm, alpha, beta, rho, Sw, Lm, traza, determinante, discriminante = construir_funciones_clasificacion(parametros)
```

#### Función para clasificar puntos fijos

Clasificamos usando las reglas del plano traza-determinante:

- Si $\delta<0$, el punto fijo es una silla.
- Si $\delta>0$ y $\tau<0$, el punto fijo es estable.
- Si $\delta>0$ y $\tau>0$, el punto fijo es inestable.
- Si $\Delta<0$, los valores propios son complejos y el punto es un foco.
- Si $\Delta>0$, los valores propios son reales y el punto es un nodo o una silla.

```python
def clasificar_punto_fijo(tau, delta, Delta, tol=1e-8):
    """
    Clasifica un punto fijo de un sistema bidimensional
    usando traza, determinante y discriminante.
    """

    if abs(delta) < tol:
        return "no hiperbólico"

    if delta < 0:
        return "silla"

    # A partir de aquí delta > 0
    if abs(tau) < tol:
        if Delta < 0:
            return "candidato a Hopf / centro lineal"
        else:
            return "no hiperbólico"

    if tau < 0:
        if Delta > 0:
            return "nodo estable"
        elif Delta < 0:
            return "foco estable"
        else:
            return "nodo degenerado estable"

    if tau > 0:
        if Delta > 0:
            return "nodo inestable"
        elif Delta < 0:
            return "foco inestable"
        else:
            return "nodo degenerado inestable"
```

### 11. Examine las bifurcaciones obtenidas en el punto anterior. Calcule trayectorias que correspondan a cada régimen de puntos fijos en un experimento de I-Clamp.

En los puntos anteriores construimos la curva de puntos fijos

$$
(J_F(v^*),v^*),
$$

y clasificamos cada punto fijo usando la traza, el determinante y el discriminante.

Ahora queremos conectar esa información con trayectorias del sistema.

La idea es la siguiente:

1. Los dobleces de la curva $(J_F,v^*)$ son candidatos a bifurcaciones tipo saddle-node.
2. Estos dobleces separan regímenes con distinto número y tipo de puntos fijos.
3. Para cada régimen elegimos un valor representativo de $J_F$.
4. Simulamos el sistema con ese valor constante de $J_F$.
5. Esto corresponde a un experimento de I-Clamp.

En un experimento de I-Clamp no fijamos el voltaje. Fijamos una corriente externa y dejamos que el voltaje evolucione libremente. Matemáticamente, eso significa que tomamos

$$
J_F(t)=J_0,
$$

constante, y resolvemos el sistema

$$
\dot v
=
J_0
-
g_US_m(v)(v-v_U)
-
g_Dw(v-v_D),
$$

$$
\dot w
=
\alpha(v)(1-w)-\beta(v)w.
$$

#### Bifurcaciones tipo silla-nodo

La curva de equilibrio está parametrizada por $v^*$:

$$
v^*
\longmapsto
(J_F(v^*),v^*).
$$

Los dobleces de esta curva ocurren cuando

$$
\frac{dJ_F}{dv^*}=0.
$$

En esos puntos, dos equilibrios se fusionan. Este es el mecanismo típico de una bifurcación saddle-node.

Por eso, para examinar las bifurcaciones, primero localizamos los valores de $v^*$ donde

$$
\frac{dJ_F}{dv^*}\approx 0.
$$

#### Localización aproximada de bifurcaciones

Usamos la curva refinada de puntos fijos. Aproximamos la derivada de $J_F(v^*)$ con diferencias finitas y buscamos cambios de signo.

```python
# Vector refinado de valores v*
vD = parametros["vD"]
vU = parametros["vU"]

v_estrellas_rama = np.linspace(vD, vU, 50000)
JF_rama = JF_equilibrio(v_estrellas_rama)
w_rama = Sw(v_estrellas_rama)

# Derivada numérica de JF respecto a v*
dJF_dv = np.gradient(JF_rama, v_estrellas_rama)

# Índices donde cambia el signo de dJF/dv
indices_dobleces = np.where(np.diff(np.sign(dJF_dv)) != 0)[0]

indices_dobleces
```

Para obtener valores más precisos de los dobleces usamos el método de Brent sobre la derivada numérica centrada.

```python
def derivada_numerica_JF(v, h=1e-5):
    """
    Aproxima dJF/dv usando diferencias centradas.
    """
    return (JF_equilibrio(v + h) - JF_equilibrio(v - h)) / (2*h)


dobleces = []

for idx in indices_dobleces:
    a = v_estrellas_rama[idx]
    b = v_estrellas_rama[idx + 1]

    try:
        raiz = brentq(derivada_numerica_JF, a, b)
        dobleces.append(raiz)
    except ValueError:
        pass

dobleces = np.array(dobleces)

tabla_dobleces = pd.DataFrame({
    "v*_doblez": dobleces,
    "JF_doblez": JF_equilibrio(dobleces),
    "w*_doblez": Sw(dobleces),
    "dJF/dv": [derivada_numerica_JF(v) for v in dobleces]
})

tabla_dobleces
```

Cada doblez separa regímenes con diferente estructura de puntos fijos.

Por tanto, elegiremos valores representativos de $J_F$ a ambos lados de los dobleces y también dentro de la región con varios puntos fijos.

$$

#### Simulación de un experimento de I-Clamp

En un experimento de I-Clamp, fijamos la corriente externa y observamos la evolución de $v(t)$.

Numéricamente, para cada régimen tomamos

$$
J_F(t)=J_0,
$$

y simulamos el sistema.

Para evitar problemas con integradores adaptativos en regímenes donde la solución puede salir rápidamente del dominio fisiológico, usaremos un método de Runge-Kutta de orden 4 con paso fijo.

Además, si la trayectoria sale demasiado del dominio, detenemos la simulación.

```python
def construir_campo_IClamp(parametros):
    """
    Construye el campo vectorial para un experimento de I-Clamp.
    """

    gU = parametros["gU"]
    gD = parametros["gD"]
    r = parametros["r"]
    vU = parametros["vU"]
    vD = parametros["vD"]
    vm = parametros["vm"]
    vw = parametros["vw"]
    sm = parametros["sm"]
    sw = parametros["sw"]
    bw = parametros["bw"]

    def exp_segura(x):
        return np.exp(np.clip(x, -700, 700))

    def Sm_local(v):
        return 1.0 / (1.0 + exp_segura(sm * (vm - v)))

    def alpha_local(v):
        return r * exp_segura(sw * bw * (v - vw))

    def beta_local(v):
        return r * exp_segura(sw * (bw - 1.0) * (v - vw))

    def campo(v, w, JF_constante):
        dv = (
            JF_constante
            - gU * Sm_local(v) * (v - vU)
            - gD * w * (v - vD)
        )

        dw = alpha_local(v) * (1.0 - w) - beta_local(v) * w

        return dv, dw

    return campo
```

```python
campo_IClamp = construir_campo_IClamp(parametros)
```

#### Integrador RK4

Implementamos una función para simular trayectorias con corriente constante.

El estado es

$$
(v(t),w(t)).
$$

La corriente $J_F$ se mantiene fija durante toda la simulación.

```python
def simular_IClamp_RK4(
    JF_constante,
    condicion_inicial,
    parametros,
    tmax=80.0,
    dt=0.01,
    limite_v=250.0
):
    """
    Simula el sistema con corriente constante JF usando RK4.
    """

    campo = construir_campo_IClamp(parametros)

    n_pasos = int(tmax / dt) + 1

    t = np.zeros(n_pasos)
    v = np.zeros(n_pasos)
    w = np.zeros(n_pasos)

    v[0], w[0] = condicion_inicial

    for k in range(n_pasos - 1):
        t[k + 1] = t[k] + dt

        vk = v[k]
        wk = w[k]

        def f_estado(v_actual, w_actual):
            dv, dw = campo(v_actual, w_actual, JF_constante)
            return np.array([dv, dw])

        yk = np.array([vk, wk])

        k1 = f_estado(yk[0], yk[1])
        k2 = f_estado(*(yk + 0.5 * dt * k1))
        k3 = f_estado(*(yk + 0.5 * dt * k2))
        k4 = f_estado(*(yk + dt * k3))

        y_siguiente = yk + (dt / 6.0) * (k1 + 2*k2 + 2*k3 + k4)

        v[k + 1] = y_siguiente[0]
        w[k + 1] = y_siguiente[1]

        # Detener si la trayectoria sale demasiado del rango fisiológico
        if abs(v[k + 1]) > limite_v or not np.isfinite(v[k + 1]) or not np.isfinite(w[k + 1]):
            t = t[:k + 2]
            v = v[:k + 2]
            w = w[:k + 2]
            break

    return t, v, w
```

### 13. Construya figuras en las que también muestre

$$
(t,v), \qquad (\partial_t v, v), \qquad (\partial_t^2 v, v).
$$


En este punto queremos complementar las gráficas del punto anterior con dos nuevas representaciones:

1. La gráfica temporal
   $$
   (t,v).
   $$
2. La gráfica de velocidad de voltaje
   $$
   (\partial_t v, v).
   $$
3. La gráfica de aceleración de voltaje
   $$
   (\partial_t^2 v, v).
   $$

La idea es mostrar estas tres gráficas juntas, compartiendo el eje vertical de $v$.

Así podemos comparar directamente:

- cómo cambia el voltaje en el tiempo,
- en qué regiones el voltaje está subiendo o bajando,
- y en qué regiones la trayectoria cambia de concavidad.

```python
def construir_funciones_dinamica_v(parametros):
    """
    Construye funciones para evaluar:
    - dot v
    - dot w
    - ddot v
    """

    gU = parametros["gU"]
    gD = parametros["gD"]
    r = parametros["r"]
    vU = parametros["vU"]
    vD = parametros["vD"]
    vm = parametros["vm"]
    vw = parametros["vw"]
    sm = parametros["sm"]
    sw = parametros["sw"]
    bw = parametros["bw"]

    def exp_segura(x):
        return np.exp(np.clip(x, -700, 700))

    def Sm(v):
        return 1.0 / (1.0 + exp_segura(sm * (vm - v)))

    def alpha(v):
        return r * exp_segura(sw * bw * (v - vw))

    def beta(v):
        return r * exp_segura(sw * (bw - 1.0) * (v - vw))

    def rho(v):
        return alpha(v) + beta(v)

    def H(v, JF_constante):
        return JF_constante - gU * Sm(v) * (v - vU)

    def K(v):
        return gD * (v - vD)

    def Lm(v):
        return 1.0 + sm * (1.0 - Sm(v)) * (v - vU)

    def Hprima(v):
        return -gU * Sm(v) * Lm(v)

    def Kprima(v):
        return gD + 0.0 * np.asarray(v)

    def dotw(v, w):
        return alpha(v) * (1.0 - w) - beta(v) * w

    def dotv(v, w, JF_constante):
        return H(v, JF_constante) - w * K(v)

    def ddotv(v, w, JF_constante):
        dv = dotv(v, w, JF_constante)
        dw = dotw(v, w)
        return (Hprima(v) - w * Kprima(v)) * dv - K(v) * dw

    return Sm, alpha, beta, rho, H, K, Hprima, Kprima, dotv, dotw, ddotv
```

### 14. Construya una serie de gráficas en las que se observe la transición de las curvas de puntos de inflexión $I_\theta$, y sea posible relacionar las bifurcaciones encontradas en el punto 11.


En los puntos anteriores obtuvimos:

1. La curva de puntos fijos:

$$
(J_F(v^*),v^*).
$$

2. Los dobleces de esta curva, que son candidatos a bifurcaciones tipo saddle-node.

3. El conjunto de puntos de inflexión:

$$
I=\{(w,v):\ddot v=0\}.
$$

4. La curva de umbrales:

$$
I_\theta=I\cap R_1.
$$

Ahora queremos observar cómo cambia $I_\theta$ cuando variamos $J_F$.

La idea será graficar varias curvas $I_\theta$ para distintos valores de corriente externa $J_F$, especialmente antes, cerca y después de los valores críticos encontrados en el punto 11.

#### Idea geométrica

La corriente externa $J_F$ cambia la forma de la ceroclina de $v$ y también cambia la curva de puntos de inflexión, porque $J_F$ aparece en la función

$$
H(v)=J_F-g_US_m(v)(v-v_U).
$$

Por tanto, al cambiar $J_F$ cambian:

- la ceroclina de voltaje $N_v$,
- las intersecciones entre ceroclinas,
- el número y tipo de puntos fijos,
- el conjunto $I=\{\ddot v=0\}$,
- y la parte relevante $I_\theta=I\cap R_1$.

En la serie de gráficas, cada panel corresponderá a un valor fijo de $J_F$.

Así podremos comparar visualmente los regímenes separados por bifurcaciones.

#### Funciones para las curvas de inflexión

Definimos funciones que, para un valor fijo de $J_F$, calculan:

- la ceroclina de $v$,
- la ceroclina de $w$,
- las ramas de $I$,
- y la parte válida de la curva de umbrales $I_\theta$.

Trabajaremos en el plano fase con ejes:

$$
x=w,
\qquad
y=v.
$$

Así el voltaje $v$ queda en el eje vertical.

```python
def construir_funciones_inflexion_para_JF(parametros, JF_constante):
    """
    Construye funciones necesarias para graficar las curvas de inflexión
    y la curva de umbrales I_theta para un valor fijo de JF.
    """

    gU = parametros["gU"]
    gD = parametros["gD"]
    r = parametros["r"]
    vU = parametros["vU"]
    vD = parametros["vD"]
    vm = parametros["vm"]
    vw = parametros["vw"]
    sm = parametros["sm"]
    sw = parametros["sw"]
    bw = parametros["bw"]

    def exp_segura(x):
        return np.exp(np.clip(x, -700, 700))

    def Sm(v):
        return 1.0 / (1.0 + exp_segura(sm * (vm - v)))

    def alpha(v):
        return r * exp_segura(sw * bw * (v - vw))

    def beta(v):
        return r * exp_segura(sw * (bw - 1.0) * (v - vw))

    def rho(v):
        return alpha(v) + beta(v)

    def Sw(v):
        return alpha(v) / rho(v)

    def H(v):
        return JF_constante - gU * Sm(v) * (v - vU)

    def K(v):
        return gD * (v - vD)

    def Nv(v):
        return H(v) / K(v)

    def Lm(v):
        return 1.0 + sm * (1.0 - Sm(v)) * (v - vU)

    def Hprima(v):
        return -gU * Sm(v) * Lm(v)

    def Kprima(v):
        return gD + 0.0 * np.asarray(v)

    def a2(v):
        return K(v) * Kprima(v)

    def a1(v):
        return K(v) * rho(v) - K(v) * Hprima(v) - H(v) * Kprima(v)

    def a0(v):
        return H(v) * Hprima(v) - K(v) * alpha(v)

    def discriminante_inflexion(v):
        return a1(v)**2 - 4.0 * a2(v) * a0(v)

    def ramas_inflexion(v):
        """
        Devuelve las dos ramas w_-(v), w_+(v).
        Donde el discriminante es negativo se coloca NaN.
        """
        D = discriminante_inflexion(v)

        w_menos = np.full_like(v, np.nan, dtype=float)
        w_mas = np.full_like(v, np.nan, dtype=float)

        mascara = (D >= 0.0) & (np.abs(a2(v)) > 1e-12)

        w_menos[mascara] = (
            -a1(v[mascara]) - np.sqrt(D[mascara])
        ) / (2.0 * a2(v[mascara]))

        w_mas[mascara] = (
            -a1(v[mascara]) + np.sqrt(D[mascara])
        ) / (2.0 * a2(v[mascara]))

        return w_menos, w_mas

    def mascara_Itheta(v, w_rama):
        """
        Selecciona la parte de una rama de I que pertenece a R1.
        """
        D = discriminante_inflexion(v)
        Nv_val = Nv(v)
        Sw_val = Sw(v)

        mascara = (
            (D >= 0.0)
            & np.isfinite(w_rama)
            & (w_rama >= 0.0)
            & (w_rama <= 1.0)
            & (w_rama < Nv_val)
            & (w_rama < Sw_val)
        )

        return mascara

    return {
        "Sm": Sm,
        "alpha": alpha,
        "beta": beta,
        "rho": rho,
        "Sw": Sw,
        "H": H,
        "K": K,
        "Nv": Nv,
        "a2": a2,
        "a1": a1,
        "a0": a0,
        "discriminante_inflexion": discriminante_inflexion,
        "ramas_inflexion": ramas_inflexion,
        "mascara_Itheta": mascara_Itheta
    }
```

#### Valores críticos de bifurcación

Los dobleces de la curva de puntos fijos se obtienen resolviendo aproximadamente

$$

Estos valores de $J_F$ separan distintos regímenes.

```python
vD = parametros["vD"]
vU = parametros["vU"]

v_estrellas_rama = np.linspace(vD, vU, 50000)
JF_rama = JF_equilibrio(v_estrellas_rama)

dJF_dv = np.gradient(JF_rama, v_estrellas_rama)
indices_dobleces = np.where(np.diff(np.sign(dJF_dv)) != 0)[0]

def derivada_numerica_JF(v, h=1e-5):
    return (JF_equilibrio(v + h) - JF_equilibrio(v - h)) / (2.0 * h)

dobleces_v = []

for idx in indices_dobleces:
    a = v_estrellas_rama[idx]
    b = v_estrellas_rama[idx + 1]

    try:
        raiz = brentq(derivada_numerica_JF, a, b)
        dobleces_v.append(raiz)
    except ValueError:
        pass

dobleces_v = np.array(dobleces_v)
dobleces_JF = JF_equilibrio(dobleces_v)
dobleces_w = Sw(dobleces_v)

tabla_dobleces = pd.DataFrame({
    "v*_doblez": dobleces_v,
    "JF_doblez": dobleces_JF,
    "w*_doblez": dobleces_w
})

tabla_dobleces
```

#### Gráfica de un panel para un valor fijo de $J_F$

La siguiente función grafica, para un valor fijo de $J_F$:

- los puntos fijos clasificados.

Usamos el plano $(w,v)$ para que el voltaje $v$ aparezca en el eje vertical.

```python
def graficar_panel_inflexion_transicion(
    ax,
    JF_constante,
    parametros,
    n_v=2500,
    mostrar_ramas_I=True,
    mostrar_puntos_fijos=True
):
    """
    Grafica un panel de transición de I_theta para un valor de JF.
    """

    funcs = construir_funciones_inflexion_para_JF(parametros, JF_constante)

    vD = parametros["vD"]
    vU = parametros["vU"]

    valores_v = np.linspace(vD + 1e-6, vU, n_v)

    Nv = funcs["Nv"](valores_v)
    Sw_val = funcs["Sw"](valores_v)

    w_menos, w_mas = funcs["ramas_inflexion"](valores_v)

    mascara_menos = funcs["mascara_Itheta"](valores_v, w_menos)
    mascara_mas = funcs["mascara_Itheta"](valores_v, w_mas)

    # Ceroclinas
    ax.plot(Nv, valores_v, linewidth=1.5, label=r"$N_v$")
    ax.plot(Sw_val, valores_v, linewidth=1.5, label=r"$N_w$")

    # Ramas completas de I, en gris
    if mostrar_ramas_I:
        mascara_I_menos = np.isfinite(w_menos) & (w_menos >= 0) & (w_menos <= 1)
        mascara_I_mas = np.isfinite(w_mas) & (w_mas >= 0) & (w_mas <= 1)

        ax.plot(
            w_menos[mascara_I_menos],
            valores_v[mascara_I_menos],
            linestyle=":",
            linewidth=1.0,
            color="0.55",
            label=r"$I$"
        )

        ax.plot(
            w_mas[mascara_I_mas],
            valores_v[mascara_I_mas],
            linestyle=":",
            linewidth=1.0,
            color="0.55"
        )

    # Curva de umbral I_theta
    ax.plot(
        w_menos[mascara_menos],
        valores_v[mascara_menos],
        color="black",
        linewidth=2.3,
        label=r"$I_\theta$"
    )

    ax.plot(
        w_mas[mascara_mas],
        valores_v[mascara_mas],
        color="black",
        linewidth=2.3
    )

    # Puntos fijos
    if mostrar_puntos_fijos:
        tabla_pf = puntos_fijos_para_corriente(JF_constante)

        marcadores = {
            "nodo estable": "o",
            "foco estable": "o",
            "silla": "s",
            "nodo inestable": "^",
            "foco inestable": "^",
            "nodo degenerado estable": "D",
            "nodo degenerado inestable": "D",
            "candidato a Hopf / centro lineal": "x",
            "no hiperbólico": "x"
        }

        for _, punto in tabla_pf.iterrows():
            ax.scatter(
                punto["w*"],
                punto["v*"],
                s=80,
                marker=marcadores.get(punto["tipo"], "o"),
                edgecolor="black",
                linewidth=1.0,
                zorder=5
            )

    ax.set_xlim(0, 1)
    ax.set_ylim(vD, vU)
    ax.grid(True)
    ax.set_title(fr"$J_F={JF_constante:.3f}$")

    return ax
```

### 15. Expanda el modelo para que, con una corriente más y una dimensión más, produzca potenciales de acción en ráfagas.



Hasta ahora trabajamos con un modelo minimal bidimensional de excitabilidad:

$$
(v,w),
$$

donde $v$ representa el potencial transmembranal y $w$ representa una variable de recuperación.

Para obtener ráfagas necesitamos que el sistema tenga una variable lenta adicional. La idea, siguiendo el enfoque de Av-Ron, Parnas y Segel, es agregar una variable asociada al calcio intracelular.

Denotaremos esta nueva variable por

$$
c(t).
$$

La variable $c$ representa, de manera reducida, la concentración intracelular de calcio. Esta concentración aumenta durante la actividad eléctrica y disminuye lentamente durante el reposo.

La nueva dimensión será entonces

$$
c.
$$

El modelo expandido tendrá variables

$$
(v,w,c).
$$

#### Intuición biofísica

La idea central es que el calcio funciona como una variable lenta que registra la actividad reciente de la neurona.

Durante una ráfaga, la neurona produce varios potenciales de acción. En cada potencial de acción entra calcio a la célula. Por tanto, la concentración intracelular de calcio $c(t)$ aumenta gradualmente.

Ese calcio intracelular puede activar canales de potasio dependientes de calcio. Cuando estos canales se abren, producen una corriente descendente que empuja el potencial de membrana hacia valores más bajos.

Así se obtiene un mecanismo natural de terminación de la ráfaga:

$$
\text{muchos potenciales de acción}
\quad\Longrightarrow\quad
\text{aumenta } c
\quad\Longrightarrow\quad
\text{aumenta corriente descendente}
\quad\Longrightarrow\quad
\text{termina la ráfaga}.
$$

Después, durante el periodo silencioso, el calcio intracelular se remueve lentamente. Al disminuir $c$, disminuye la corriente descendente dependiente de calcio, y la neurona puede volver a estar lista para disparar otra ráfaga.

Por tanto, el mecanismo cualitativo es:

$$
\text{actividad rápida}
\longrightarrow
\text{acumulación lenta de calcio}
\longrightarrow
\text{adaptación}
\longrightarrow
\text{silencio}
\longrightarrow
\text{recuperación}
\longrightarrow
\text{nueva ráfaga}.
$$

#### Corriente adicional

Agregamos una corriente de potasio dependiente de calcio:

$$
J_{KCa}(c,v)
=
g_{KCa}\frac{c}{c+K_d}(v-v_D).
$$

Aquí:

- $g_{KCa}>0$ es la conductancia máxima de la corriente dependiente de calcio;
- $K_d>0$ es una constante de semisaturación;
- $c$ es la concentración intracelular de calcio;
- $v_D$ es el potencial de reversa de la corriente descendente.

La función

$$
\frac{c}{c+K_d}
$$

es creciente y saturante. Esto significa que, cuando hay poco calcio, la corriente es pequeña; cuando el calcio aumenta, la corriente crece, pero no indefinidamente.

Como $v_D$ es un potencial bajo, si $v>v_D$, entonces

$$
J_{KCa}(c,v)>0.
$$

En la ecuación de voltaje esta corriente aparece restando:

$$
-J_{KCa}(c,v).
$$

Por tanto, al aumentar $c$, aumenta una corriente que baja el voltaje. Esta es la adaptación lenta que permite terminar una ráfaga.

#### Entrada de calcio

Para que $c(t)$ aumente durante los potenciales de acción, necesitamos modelar una entrada de calcio.

Siguiendo la idea de Av-Ron, podemos suponer que el calcio entra durante la despolarización. Una forma simple es introducir una corriente de calcio de la forma

$$
J_{Ca}(v,w)
=
g_{Ca}S_m(v)(1-w)(v-v_{Ca}).
$$

Aquí:

- $g_{Ca}>0$ controla la intensidad de la entrada de calcio;
- $v_{Ca}$ es el potencial de reversa del calcio, típicamente alto;
- $S_m(v)$ representa activación rápida por voltaje;
- el factor $(1-w)$ hace que la entrada sea mayor cuando la recuperación todavía no ha cerrado la fase ascendente.

Como normalmente

$$
v<v_{Ca},
$$

entonces

$$
v-v_{Ca}<0.
$$

Por tanto,

$$
J_{Ca}(v,w)<0.
$$

Es decir, $J_{Ca}$ es una corriente entrante.

Para traducir corriente entrante en aumento de calcio intracelular usamos

$$
-K_pJ_{Ca}(v,w),
$$

donde $K_p>0$ convierte corriente en concentración. Como $J_{Ca}<0$, entonces

$$
-K_pJ_{Ca}>0.
$$

Por eso la corriente de calcio aumenta $c$.

#### Dinámica del calcio

La nueva variable $c(t)$ satisface

$$
\dot c
=
-K_pJ_{Ca}(v,w)-Rc.
$$

El primer término representa entrada de calcio:

$$
-K_pJ_{Ca}(v,w).
$$

El segundo término representa remoción o bombeo de calcio:

$$
-Rc.
$$

Aquí:

- $K_p>0$ es un factor de conversión de corriente a concentración;
- $R>0$ es la tasa de remoción de calcio.

Durante una ráfaga, $J_{Ca}$ es entrante y $c$ aumenta.

Durante el silencio, la entrada de calcio disminuye y domina el término $-Rc$, por lo que $c$ decae lentamente.

#### Modelo expandido

El modelo bidimensional original se expande a un modelo tridimensional agregando la variable $c$ y la corriente dependiente de calcio.

El sistema propuesto es

$$
\boxed{
\dot v
=
J_F
-
g_US_m(v)(v-v_U)
-
g_Dw(v-v_D)
-
g_{KCa}\frac{c}{c+K_d}(v-v_D)
}
$$

$$
\boxed{
\dot w
=
\alpha(v)(1-w)-\beta(v)w
}
$$

$$
\boxed{
\dot c
=
-K_pJ_{Ca}(v,w)-Rc
}
$$

donde

$$
\boxed{
J_{Ca}(v,w)
=
g_{Ca}S_m(v)(1-w)(v-v_{Ca}).
}
$$

Este modelo tiene una dimensión adicional:

$$
c,
$$

y una corriente adicional de adaptación dependiente de calcio:

$$
J_{KCa}(c,v)
=
g_{KCa}\frac{c}{c+K_d}(v-v_D).
$$

La corriente de calcio $J_{Ca}$ alimenta la dinámica de $c$, es decir, permite que la variable lenta aumente durante la actividad eléctrica.

### 16. Encuentre soluciones que demuestren que el modelo produce ráfagas.


En el punto anterior expandimos el modelo minimal Up-Down agregando una variable lenta de calcio intracelular $c(t)$ y una corriente de potasio dependiente de calcio.

El objetivo ahora es encontrar soluciones numéricas que muestren ráfagas, es decir, grupos de potenciales de acción separados por periodos de silencio.

La estructura que queremos observar es:

$$
\text{silencio}
\longrightarrow
\text{varios potenciales de acción}
\longrightarrow
\text{silencio}
\longrightarrow
\text{nueva ráfaga}.
$$

La variable lenta $c(t)$ debe aumentar durante la fase activa y disminuir durante la fase silenciosa.

#### Parámetros

Usamos los parámetros del modelo Up-Down y agregamos parámetros para el calcio.

Los parámetros más sensibles para obtener ráfagas son:

- $J_F$, que controla la excitación basal;
- $g_{KCa}$, que controla la fuerza de la corriente adaptativa;
- $K_p$, que controla cuánto aumenta $c$ por entrada de calcio;
- $R$, que controla qué tan rápido se remueve el calcio.


```python
parametros_burst_adaptado = {
    # Modelo Up-Down original
    "gU": 2.0,
    "gD": 10.0,
    "r": 0.1,
    "vU": 60.0,
    "vD": -80.0,
    "vm": -20.0,
    "vw": 0.0,
    "sm": 0.22,
    "sw": 0.11,
    "bw": 0.35,

    # Corriente externa
    "JF": 20.0,

    # Corriente KCa
    "gKCa": 6.0,
    "Kd": 0.5,

    # Corriente de calcio que alimenta c(t)
    "gCa": 0.5,
    "vCa": 120.0,

    # Dinámica del calcio
    "Kp": 1e-4,
    "R": 0.01
}
```

```python
def construir_modelo_burst_adaptado(parametros):
    """
    Construye el campo vectorial del modelo Up-Down expandido con calcio.

    Estado:
        y = (v, w, c)

    donde:
        v : potencial de membrana
        w : variable de recuperación
        c : calcio intracelular
    """

    gU = parametros["gU"]
    gD = parametros["gD"]
    r = parametros["r"]

    vU = parametros["vU"]
    vD = parametros["vD"]

    vm = parametros["vm"]
    vw = parametros["vw"]

    sm = parametros["sm"]
    sw = parametros["sw"]
    bw = parametros["bw"]

    JF = parametros["JF"]

    gKCa = parametros["gKCa"]
    Kd = parametros["Kd"]

    gCa = parametros["gCa"]
    vCa = parametros["vCa"]

    Kp = parametros["Kp"]
    R = parametros["R"]

    def exp_segura(x):
        return np.exp(np.clip(x, -60, 60))

    def Sm(v):
        return 1.0 / (1.0 + exp_segura(sm * (vm - v)))

    def alpha(v):
        return r * exp_segura(sw * bw * (v - vw))

    def beta(v):
        return r * exp_segura(sw * (bw - 1.0) * (v - vw))

    def corrientes(v, w, c):
        c = np.maximum(c, 0.0)

        JU = gU * Sm(v) * (v - vU)
        JD = gD * w * (v - vD)

        JKCa = gKCa * (c / (c + Kd)) * (v - vD)

        JCa = gCa * Sm(v) * (1.0 - w) * (v - vCa)

        return JU, JD, JKCa, JCa

    def campo(y):
        v, w, c = y
        c = max(c, 0.0)

        JU, JD, JKCa, JCa = corrientes(v, w, c)

        dv = JF - JU - JD - JKCa
        dw = alpha(v) * (1.0 - w) - beta(v) * w
        dc = -Kp * JCa - R * c

        return np.array([dv, dw, dc])

    return campo, corrientes, Sm, alpha, beta
```

```python
def simular_burst_adaptado_RK4(
    parametros,
    condicion_inicial=(-75.0, 0.001, 0.0),
    tmax=1500.0,
    dt=0.02,
    limite_v=300.0
):
    """
    Simula el modelo adaptado usando RK4.
    """

    campo, corrientes, Sm, alpha, beta = construir_modelo_burst_adaptado(parametros)

    n_pasos = int(tmax / dt) + 1

    t = np.zeros(n_pasos)
    v = np.zeros(n_pasos)
    w = np.zeros(n_pasos)
    c = np.zeros(n_pasos)

    v[0], w[0], c[0] = condicion_inicial

    for k in range(n_pasos - 1):
        t[k + 1] = t[k] + dt

        y = np.array([v[k], w[k], c[k]])

        k1 = campo(y)
        k2 = campo(y + 0.5 * dt * k1)
        k3 = campo(y + 0.5 * dt * k2)
        k4 = campo(y + dt * k3)

        y_siguiente = y + (dt / 6.0) * (k1 + 2*k2 + 2*k3 + k4)

        v[k + 1] = y_siguiente[0]
        w[k + 1] = y_siguiente[1]
        c[k + 1] = max(y_siguiente[2], 0.0)

        if (
            abs(v[k + 1]) > limite_v
            or not np.isfinite(v[k + 1])
            or not np.isfinite(w[k + 1])
            or not np.isfinite(c[k + 1])
        ):
            t = t[:k + 2]
            v = v[:k + 2]
            w = w[:k + 2]
            c = c[:k + 2]
            break

    return t, v, w, c
```

#### Simulación de una solución

Simulamos el sistema desde una condición inicial cerca del reposo.

```python
t_burst, v_burst, w_burst, c_burst = simular_burst_adaptado_RK4(
    parametros=parametros_burst_adaptado,
    condicion_inicial=(-75.0, 0.001, 0.0),
    tmax=1500.0,
    dt=0.02
)
```

### 17. Jacobiana del modelo tridimensional adaptado

Consideramos el modelo expandido

$$
\dot v
=
J_F
-
g_US_m(v)(v-v_U)
-
g_Dw(v-v_D)
-
g_{KCa}\frac{c}{c+K_d}(v-v_D),
$$

$$
\dot w
=
\alpha(v)(1-w)-\beta(v)w,
$$

$$
\dot c
=
-K_pg_{Ca}S_m(v)(1-w)(v-v_{Ca})-Rc.
$$

Trabajaremos con el orden de variables

$$
(v,w,c).
$$

La matriz Jacobiana será

$$
J(v,w,c)
=
\begin{pmatrix}
\partial_v\dot v & \partial_w\dot v & \partial_c\dot v\\
\partial_v\dot w & \partial_w\dot w & \partial_c\dot w\\
\partial_v\dot c & \partial_w\dot c & \partial_c\dot c
\end{pmatrix}.
$$

#### Derivadas auxiliares

Recordemos que

$$
S_m(v)
=
\frac{1}{1+\exp[s_m(v_m-v)]}.
$$

Por tanto,

$$
S_m'(v)
=
s_mS_m(v)(1-S_m(v)).
$$

Además,

$$
\alpha(v)
=
r\exp[s_wb_w(v-v_w)],
$$

$$
\beta(v)
=
r\exp[s_w(b_w-1)(v-v_w)].
$$

Entonces

$$
\alpha'(v)
=
s_wb_w\alpha(v),
$$

y

$$
\beta'(v)
=
s_w(b_w-1)\beta(v).
$$

También definimos

$$
\rho(v)=\alpha(v)+\beta(v),
$$

de modo que

$$
\dot w=\alpha(v)-w\rho(v).
$$

Finalmente, para la corriente dependiente de calcio usamos

$$
Q(c)=\frac{c}{c+K_d}.
$$

Entonces

$$
Q'(c)=\frac{K_d}{(c+K_d)^2}.
$$

#### Primera fila de la Jacobiana

La ecuación de voltaje es

$$
\dot v
=
J_F
-
g_US_m(v)(v-v_U)
-
g_Dw(v-v_D)
-
g_{KCa}Q(c)(v-v_D).
$$

Derivando respecto a $v$:

$$
\partial_v\dot v
=
-g_U\left[
S_m'(v)(v-v_U)+S_m(v)
\right]
-g_Dw
-g_{KCa}Q(c).
$$

Derivando respecto a $w$:

$$
\partial_w\dot v
=
-g_D(v-v_D).
$$

Derivando respecto a $c$:

$$
\partial_c\dot v
=
-g_{KCa}Q'(c)(v-v_D).
$$

Como

$$
Q'(c)=\frac{K_d}{(c+K_d)^2},
$$

obtenemos

$$
\partial_c\dot v
=
-g_{KCa}
\frac{K_d}{(c+K_d)^2}
(v-v_D).
$$

#### Segunda fila de la Jacobiana

La ecuación de recuperación es

$$
\dot w
=
\alpha(v)(1-w)-\beta(v)w.
$$

Equivalente a

$$
\dot w=\alpha(v)-w\rho(v).
$$

Derivando respecto a $v$:

$$
\partial_v\dot w
=
\alpha'(v)-w\rho'(v).
$$

Como

$$
\rho'(v)=\alpha'(v)+\beta'(v),
$$

también podemos escribir

$$
\partial_v\dot w
=
\alpha'(v)(1-w)-\beta'(v)w.
$$

Usando las derivadas explícitas,

$$
\partial_v\dot w
=
s_wb_w\alpha(v)(1-w)
-
s_w(b_w-1)\beta(v)w.
$$

Derivando respecto a $w$:

$$
\partial_w\dot w
=
-\alpha(v)-\beta(v)
=
-\rho(v).
$$

Derivando respecto a $c$:

$$
\partial_c\dot w=0.
$$

#### Tercera fila de la Jacobiana

La ecuación del calcio es

$$
\dot c
=
-K_pg_{Ca}S_m(v)(1-w)(v-v_{Ca})-Rc.
$$

Derivando respecto a $v$:

$$
\partial_v\dot c
=
-K_pg_{Ca}(1-w)
\left[
S_m'(v)(v-v_{Ca})+S_m(v)
\right].
$$

Derivando respecto a $w$:

$$
\partial_w\dot c
=
K_pg_{Ca}S_m(v)(v-v_{Ca}).
$$

Derivando respecto a $c$:

$$
\partial_c\dot c
=
-R.
$$

#### Jacobiana final

Por tanto, la matriz Jacobiana del modelo tridimensional adaptado es

$$
\boxed{
J(v,w,c)
=
\begin{pmatrix}
-g_U\left[S_m'(v)(v-v_U)+S_m(v)\right]-g_Dw-g_{KCa}\dfrac{c}{c+K_d}
&
-g_D(v-v_D)
&
-g_{KCa}\dfrac{K_d}{(c+K_d)^2}(v-v_D)
\\[12pt]
\alpha'(v)-w\rho'(v)
&
-\rho(v)
&
0
\\[12pt]
-K_pg_{Ca}(1-w)\left[S_m'(v)(v-v_{Ca})+S_m(v)\right]
&
K_pg_{Ca}S_m(v)(v-v_{Ca})
&
-R
\end{pmatrix}.
}
$$

Usando

$$
S_m'(v)=s_mS_m(v)(1-S_m(v)),
$$

$$
\alpha'(v)=s_wb_w\alpha(v),
$$

$$
\beta'(v)=s_w(b_w-1)\beta(v),
$$

y

$$
\rho'(v)=\alpha'(v)+\beta'(v).
$$
