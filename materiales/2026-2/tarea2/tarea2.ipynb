## Ecuaciones (1)--(6): energía de transporte transmembranal

El objetivo de esta primera parte es entender cómo se construye la expresión para la energía requerida al transportar moléculas a través de una membrana. La idea central es que el transporte depende de dos contribuciones: una contribución química, asociada con las diferencias de concentración, y una contribución eléctrica, asociada con el movimiento de cargas a través de una diferencia de potencial.

En el artículo se consideran dos compartimentos:

$$
0 = \text{compartimento extracelular},
$$

$$
1 = \text{compartimento intracelular}.
$$

Para una especie molecular $s$, por ejemplo $Na^+$, $K^+$, $Ca^{2+}$, $Cl^-$ o glucosa, se denotan por $c_s$ y $d_s$ los compartimentos fuente y destino, respectivamente. Es decir,

$$
c_s = \text{compartimento fuente},
$$

$$
d_s = \text{compartimento destino}.
$$

Por lo tanto,

$$
(c_s,d_s)=(0,1)
$$

representa transporte hacia adentro de la célula, mientras que

$$
(c_s,d_s)=(1,0)
$$

representa transporte hacia afuera de la célula.

De esta convención se obtiene una observación importante:

$$
c_s-d_s=
\begin{cases}
-1, & \text{si el transporte es hacia adentro},\\
1, & \text{si el transporte es hacia afuera}.
\end{cases}
$$

Este signo será fundamental en todos los cálculos posteriores, porque determina si el transporte ocurre a favor o en contra del gradiente electroquímico.

### Energía libre y energía electroquímica

La cantidad $\Delta G_s$ representa la energía libre asociada con transportar $n_s$ moléculas de tipo $s$ desde el compartimento $c_s$ hasta el compartimento $d_s$.

En este contexto, la energía libre mide si un transporte puede ocurrir espontáneamente o si requiere una fuente externa de energía.

Si

$$
\Delta G_s<0,
$$

entonces el transporte es termodinámicamente favorable. Esto significa que puede ocurrir pasivamente, disminuyendo el gradiente electroquímico.

Si

$$
\Delta G_s>0,
$$

entonces el transporte no es favorable en esa dirección y requiere energía externa, por ejemplo energía proveniente de la hidrólisis de ATP.

Aunque el artículo no introduce explícitamente el potencial electroquímico $\mu$, podemos usarlo como herramienta auxiliar para entender de dónde viene la ecuación (1). La energía electroquímica de una molécula de tipo $s$ en el compartimento $j$ puede escribirse como

$$
\mu_{s,j}
=
\mu_s^0
+
kT\ln([s]_j)
+
z_s q U_j.
$$

Aquí, $\mu_s^0$ es una energía de referencia para la especie $s$, $k$ es la constante de Boltzmann, $T$ es la temperatura absoluta, $[s]_j$ es la concentración de la especie $s$ en el compartimento $j$, $z_s$ es la valencia de la especie, $q$ es la carga elemental y $U_j$ es el potencial eléctrico en el compartimento $j$.

El subíndice $j$ indica el compartimento:

$$
j=0 \quad \text{extracelular},
$$

$$
j=1 \quad \text{intracelular}.
$$

Así, $\mu_{s,0}$ representa la energía electroquímica de la especie $s$ en el exterior, mientras que $\mu_{s,1}$ representa la energía electroquímica de la especie $s$ en el interior.

### Cálculo a mano 1: diferencia de energía electroquímica

Para transportar $n_s$ moléculas desde $c_s$ hasta $d_s$, calculamos la diferencia

$$
\Delta G_s
=
n_s(\mu_{s,d_s}-\mu_{s,c_s}).
$$

Sustituyendo la expresión del potencial electroquímico,

$$
\Delta G_s
=
n_s
\left[
\mu_s^0
+
kT\ln([s]_{d_s})
+
z_s q U_{d_s}
-
\left(
\mu_s^0
+
kT\ln([s]_{c_s})
+
z_s q U_{c_s}
\right)
\right].
$$

La energía de referencia se cancela:

$$
\mu_s^0-\mu_s^0=0.
$$

Entonces,

$$
\Delta G_s
=
n_s
\left[
kT\ln\left(\frac{[s]_{d_s}}{[s]_{c_s}}\right)
+
z_s q (U_{d_s}-U_{c_s})
\right].
$$

Este cálculo es importante hacerlo a mano porque muestra que la energía de transporte tiene dos partes: una parte química y una parte eléctrica.

La parte química es

$$
kT\ln\left(\frac{[s]_{d_s}}{[s]_{c_s}}\right),
$$

y la parte eléctrica es

$$
z_s q (U_{d_s}-U_{c_s}).
$$

### Cálculo a mano 2: reescritura usando los compartimentos 0 y 1

El artículo usa las concentraciones $[s]_0$ y $[s]_1$, correspondientes a los compartimentos extracelular e intracelular. Queremos reescribir

$$
\ln\left(\frac{[s]_{d_s}}{[s]_{c_s}}\right)
$$

en términos de

$$
\ln\left(\frac{[s]_0}{[s]_1}\right).
$$

Como $c_s,d_s\in\{0,1\}$, se cumple que

$$
\ln\left(\frac{[s]_{d_s}}{[s]_{c_s}}\right)
=
(c_s-d_s)\ln\left(\frac{[s]_0}{[s]_1}\right).
$$

Verifiquemos los dos casos.

Si el transporte es hacia adentro, entonces $(c_s,d_s)=(0,1)$ y

$$
c_s-d_s=-1.
$$

Por tanto,

$$
(c_s-d_s)\ln\left(\frac{[s]_0}{[s]_1}\right)
=
-\ln\left(\frac{[s]_0}{[s]_1}\right)
=
\ln\left(\frac{[s]_1}{[s]_0}\right),
$$

que coincide con

$$
\ln\left(\frac{[s]_{d_s}}{[s]_{c_s}}\right)
=
\ln\left(\frac{[s]_1}{[s]_0}\right).
$$

Si el transporte es hacia afuera, entonces $(c_s,d_s)=(1,0)$ y

$$
c_s-d_s=1.
$$

Por tanto,

$$
(c_s-d_s)\ln\left(\frac{[s]_0}{[s]_1}\right)
=
\ln\left(\frac{[s]_0}{[s]_1}\right),
$$

que coincide con

$$
\ln\left(\frac{[s]_{d_s}}{[s]_{c_s}}\right)
=
\ln\left(\frac{[s]_0}{[s]_1}\right).
$$

Ahora consideramos el término eléctrico. El potencial transmembranal se define como

$$
v=U_1-U_0.
$$

Queremos escribir $U_{d_s}-U_{c_s}$ en términos de $v$. Se cumple que

$$
U_{d_s}-U_{c_s}=-(c_s-d_s)v.
$$

De nuevo, podemos verificar los dos casos.

Si el transporte es hacia adentro, $(c_s,d_s)=(0,1)$, entonces

$$
U_{d_s}-U_{c_s}=U_1-U_0=v.
$$

Como $c_s-d_s=-1$,

$$
-(c_s-d_s)v=-(-1)v=v.
$$

Si el transporte es hacia afuera, $(c_s,d_s)=(1,0)$, entonces

$$
U_{d_s}-U_{c_s}=U_0-U_1=-v.
$$

Como $c_s-d_s=1$,

$$
-(c_s-d_s)v=-v.
$$

Por tanto, en ambos casos,

$$
U_{d_s}-U_{c_s}=-(c_s-d_s)v.
$$

Sustituyendo estas dos identidades en la expresión de $\Delta G_s$, obtenemos

$$
\Delta G_s
=
n_s
\left[
kT(c_s-d_s)\ln\left(\frac{[s]_0}{[s]_1}\right)
-
z_s q(c_s-d_s)v
\right].
$$

Factorizando $n_s(c_s-d_s)$,

$$
\Delta G_s
=
n_s(c_s-d_s)
\left[
kT\ln\left(\frac{[s]_0}{[s]_1}\right)
-
z_s qv
\right].
$$

Esta es la ecuación (1):

$$
\boxed{
\Delta G_s
=
n_s(c_s-d_s)
\left[
kT\ln\left(\frac{[s]_0}{[s]_1}\right)
-
z_s qv
\right].
}
$$

### Cálculo a mano 3: caso iónico y potencial de Nernst

Si $s$ es un ion, entonces $z_s\neq 0$. En este caso, se define el potencial de Nernst de la especie $s$ como

$$
v_s
=
\frac{kT}{z_s q}
\ln\left(\frac{[s]_0}{[s]_1}\right).
$$

Multiplicando ambos lados por $z_s q$, obtenemos

$$
z_s q v_s
=
kT\ln\left(\frac{[s]_0}{[s]_1}\right).
$$

Sustituimos esto en la ecuación (1):

$$
\Delta G_s
=
n_s(c_s-d_s)
\left[
z_s q v_s
-
z_s qv
\right].
$$

Factorizando $z_s q$,

$$
\Delta G_s
=
z_s q n_s(c_s-d_s)(v_s-v).
$$

Esta es la ecuación (2):

$$
\boxed{
\Delta G_s
=
z_s q n_s(c_s-d_s)(v_s-v).
}
$$

Esta expresión muestra que, para iones, la energía de transporte depende de la diferencia entre el potencial de Nernst $v_s$ y el potencial transmembranal $v$.

### Ejemplo breve: transporte de $K^+$ hacia afuera

Para un ion $K^+$ que se transporta hacia afuera, tenemos

$$
z_K=1,
\qquad
n_K=1,
\qquad
(c_K,d_K)=(1,0).
$$

Entonces

$$
c_K-d_K=1.
$$

Usando la ecuación (2),

$$
\Delta G_K
=
q(v_K-v).
$$

Este ejemplo es útil porque muestra cómo la dirección del transporte queda codificada en el factor $c_s-d_s$.

### Cálculo a mano 4: caso de moléculas no cargadas

Si la especie $s$ no tiene carga, entonces

$$
z_s=0.
$$

En ese caso, el término eléctrico desaparece de la ecuación (1). Por tanto,

$$
\Delta G_s
=
n_s(c_s-d_s)
kT\ln\left(\frac{[s]_0}{[s]_1}\right).
$$

Esta es la ecuación (3):

$$
\boxed{
\Delta G_s
=
n_s(c_s-d_s)
kT\ln\left(\frac{[s]_0}{[s]_1}\right).
}
$$

En este caso, el transporte depende únicamente del gradiente químico, es decir, de la diferencia de concentraciones entre los dos compartimentos.

### Transporte conjunto de varias especies

Ahora supongamos que un mecanismo transporta varias especies en un solo evento. Por ejemplo, un intercambiador puede transportar simultáneamente $Na^+$ y $Ca^{2+}$.

Sea $S$ el conjunto de especies transportadas en un evento. Entonces la energía total requerida es la suma de las energías individuales:

$$
\boxed{
\Delta G_S
=
\sum_{s\in S}\Delta G_s.
}
$$

Esta es la ecuación (4).

Por ejemplo, si

$$
S=\{Na^+,Ca^{2+}\},
$$

entonces

$$
\Delta G_S
=
\Delta G_{Na}
+
\Delta G_{Ca}.
$$

### Inclusión de una fuente externa de energía

Si

$$
\Delta G_S\leq 0,
$$

el transporte conjunto es termodinámicamente favorable. En cambio, si

$$
\Delta G_S>0,
$$

el transporte requiere energía externa.

Para incluir esta posibilidad, el artículo define

$$
\boxed{
\Delta G
=
\delta_{\mathrm{Ext}}\Delta G_{\mathrm{Ext}}
+
\Delta G_S.
}
$$

Esta es la ecuación (5). Aquí, $\Delta G_{\mathrm{Ext}}$ representa la energía externa disponible, por ejemplo la energía de hidrólisis de ATP, y

$$
\delta_{\mathrm{Ext}}
=
\begin{cases}
1, & \text{si } \Delta G_S>0,\\
0, & \text{si } \Delta G_S\leq 0.
\end{cases}
$$

La interpretación es la siguiente. Si el transporte ya es favorable, no hace falta energía externa y $\delta_{\mathrm{Ext}}=0$. Si el transporte no es favorable, entonces $\delta_{\mathrm{Ext}}=1$ y se agrega una fuente externa de energía.

Para que el transporte activo sea posible, la energía total debe satisfacer

$$
\Delta G\leq 0.
$$

Es decir, si $\Delta G_S>0$, entonces $\Delta G_{\mathrm{Ext}}$ debe ser suficientemente negativa para compensar el costo energético del transporte.

### Energía de hidrólisis de ATP

En transporte activo primario, una fuente común de energía es la hidrólisis de ATP. El artículo escribe esta energía como

$$
\boxed{
\Delta G_{\mathrm{ATP}}
=
\Delta G_{\mathrm{ATP}}^0
+
kT\ln\left(
\frac{[\mathrm{ADP}]_1[\mathrm{P_i}]_1}{[\mathrm{ATP}]_1}
\right)
=
qv_{\mathrm{ATP}}.
}
$$

Esta es la ecuación (6).

Aquí, $\Delta G_{\mathrm{ATP}}^0$ es el cambio estándar de energía libre de la hidrólisis de ATP, mientras que $[\mathrm{ATP}]_1$, $[\mathrm{ADP}]_1$ y $[\mathrm{P_i}]_1$ son concentraciones intracelulares.

La cantidad $v_{\mathrm{ATP}}$ es un potencial equivalente asociado con la energía libre de hidrólisis del ATP. Esta escritura permite expresar la energía química del ATP en unidades comparables con las energías eléctricas de transporte iónico.

### Comentario pedagógico

Los cálculos más importantes para hacer a mano son:

1. La cancelación de $\mu_s^0$ al calcular $\mu_{s,d_s}-\mu_{s,c_s}$.
2. La verificación de la identidad

$$
\ln\left(\frac{[s]_{d_s}}{[s]_{c_s}}\right)
=
(c_s-d_s)\ln\left(\frac{[s]_0}{[s]_1}\right).
$$

3. La verificación de la identidad

$$
U_{d_s}-U_{c_s}=-(c_s-d_s)v.
$$

4. La sustitución del potencial de Nernst

$$
v_s
=
\frac{kT}{z_s q}
\ln\left(\frac{[s]_0}{[s]_1}\right)
$$

para obtener

$$
\Delta G_s
=
z_s q n_s(c_s-d_s)(v_s-v).
$$

Estos pasos son los que realmente consolidan la comprensión de la ecuación (1). La inteligencia artificial puede ayudar a organizar la derivación y revisar signos, pero escribir estos pasos a mano obliga a seguir la lógica del modelo: dirección del transporte, gradiente químico, contribución eléctrica y energía libre.
