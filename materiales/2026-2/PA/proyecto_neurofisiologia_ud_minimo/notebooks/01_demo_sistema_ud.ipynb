# Demo mínima del sistema UD-básico

Este notebook solo importa el paquete, carga un caché NPZ ya construida y crea el widget interactivo.

```python
%matplotlib widget
from pathlib import Path
import sys

CANDIDATOS = [Path.cwd(), Path.cwd().parent]
RAIZ = None
for base in CANDIDATOS:
    if (base / 'src').exists() and (base / 'data').exists():
        RAIZ = base
        break
if RAIZ is None:
    raise RuntimeError('No se pudo inferir la raíz del proyecto.')
SRC = RAIZ / 'src'
if str(SRC) not in sys.path:
    sys.path.append(str(SRC))
```

```python
from neurofisiologia_ud import AlmacenNPZ, ParametrosUDBasico, SistemaUDBasico
from neurofisiologia_ud.visualizacion import VisualizadorUmbralInteractivo

ruta_cache = RAIZ / 'data' / 'cache' / 'cache_umbral_ejemplo.npz'
cache = AlmacenNPZ.cargar_cache_umbral(ruta_cache)
modelo = SistemaUDBasico(ParametrosUDBasico())
visualizador = VisualizadorUmbralInteractivo(cache=cache, modelo=modelo)
visualizador.construir()
```
